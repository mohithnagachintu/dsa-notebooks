### Easy

##### 1. Missing Ranges

```python
class Solution:
    def findMissingRanges(self, nums: List[int], lower: int, upper: int) -> List[List[int]]:
        n = len(nums)
        if n == 0:
            return [[lower,upper]]
        ans = []
        if lower < nums[0]:
            ans.append([lower,nums[0]-1])
        for i in range(1,n):
            if nums[i] - nums[i-1] > 1:
                ans.append([nums[i-1]+1,nums[i]-1])
        if upper > nums[-1]:
            ans.append([nums[-1]+1,upper])
        return ans
```

##### 3. Meeting Rooms - I

```python
class Solution:
    def canAttendMeetings(self, intervals: List[List[int]]) -> bool:
        intervals.sort(key = (lambda x : (x[0],x[1])))
        for i in range(1, len(intervals)):
            if intervals[i][0] < intervals[i-1][1]:
                return False
        return True
```

##### 2. Merge Intervals

```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        n = len(intervals)
        intervals.sort()
        ans = [intervals[0]]
        
        for i in range(1,n):
            if intervals[i][0] > ans[-1][1]:
                ans.append(intervals[i])
            else:
                ans[-1][1] = max(intervals[i][1],ans[-1][1])
        return ans
```

##### 3. Non overlapping Intervals

```python
class Solution:
    def eraseOverlapIntervals(self, intervals: List[List[int]]) -> int:
        # Standard Problem
        # minimize the intervals to be removed
        # smaller intervals with early end time should be prioritized

        intervals.sort(key = (lambda x: x[1]))
        ans = 0
        temp = intervals[0][1]
        for i in range(1,len(intervals)):
            if intervals[i][0] < temp:
                ans += 1
            else:
                temp = intervals[i][1]
        return ans
```

##### 4. Interval List Intersections

```python
class Solution:
    def intervalIntersection(self, firstList: List[List[int]], secondList: List[List[int]]) -> List[List[int]]:
        n,m  = len(firstList), len(secondList)
        ans = []
    
        i,j = 0,0
        while i < n and j < m:
            start = max(firstList[i][0], secondList[j][0])
            end = min(firstList[i][1], secondList[j][1])

            if start <= end:
                ans.append([start,end])
            
            if firstList[i][1] < secondList[j][1]:
                i += 1
            else: 
                j +=1
        return ans
```

##### 4. Meeting Rooms II (Important)

```python
from heapq import heappush, heappop
class Solution:
    def minMeetingRooms(self, intervals: List[List[int]]) -> int:
        intervals.sort(key=(lambda x: (x[0], x[1])))

        min_heap = []
        for start, end in intervals:
            if min_heap and start >= min_heap[0]:
                # reuse same room
                heappop(min_heap)
            # Every meeting - a room
            heappush(min_heap, end)
        return len(min_heap)
```

##### 5. Divide Intervals Into Minimum Number of Groups

```python
class Solution:
    def minGroups(self, intervals: List[List[int]]) -> int:
        # Same as Meeting Rooms II question. 

        intervals.sort(key=(lambda x: (x[0], x[1])))

        min_heap = []
        for start, end in intervals:
            if min_heap and start > min_heap[0]:
                # reuse same room
                heappop(min_heap)
            # Every meeting - a room
            heappush(min_heap, end)
        return len(min_heap)
```

##### 6. Remove Covered Intervals

```python
class Solution:
    def removeCoveredIntervals(self, intervals: List[List[int]]) -> int:
        intervals.sort(key = (lambda x: (x[0], -x[1])))

        n = len(intervals)
        temp = intervals[0]
        ans  = n
        for i in range(1,n):
            start,end = intervals[i]
            if temp[0] <= start and temp[1] >= end:
                ans -= 1
            else:
                temp = intervals[i]
        return ans

        # TC = O(N log N)
        # SC = O(1)
```

##### 6. Minimum Number of Arrows to Burst Balloons

```python
class Solution:
    def findMinArrowShots(self, points: List[List[int]]) -> int:
        points.sort(key = (lambda x: (x[0],-x[1])))
        n = len(points)
        temp = points[0]
        ans = 0
        for i in range(1,n):
            if points[i][0] <= temp[1]:
                temp[1] = min(temp[1], points[i][1])
            else:
                temp = points[i]
                ans += 1
        return ans + 1
```

##### 7. The Number of the Smallest Unoccupied Chair


```python
from heapq import heappush, heappop
class Solution:
    def smallestChair(self, times: List[List[int]], targetFriend: int) -> int:
        times = [(start, end, i) for i, (start, end) in enumerate(times)]
        times.sort()
        
        unused_heap = []
        next_chair = 0
        min_heap = []

        for start,end,friend in times:
            # free up chairs
            while min_heap and start >= min_heap[0][0]:
                _, chair = heappop(min_heap)
                heappush(unused_heap, chair)
        
            if unused_heap:
                chair = heappop(unused_heap)
            else:
                chair = next_chair
                next_chair += 1

            heappush(min_heap, (end, chair))
            
            if friend == targetFriend:
                return chair
```

##### 8. Check if Grid can be Cut into Sections (Good Pattern)

```python
class Solution:
    def checkValidCuts(self, n: int, rectangles: List[List[int]]) -> bool:
        # merge x or y and check if the size >= 3
        x_range, y_range = [],[]
        for a,b,c,d in rectangles:
            x_range.append([a,c])
            y_range.append([b,d])


        def merge_intervals(intervals):
            intervals.sort()
            ans = []
            temp = intervals[0]
            for i in range(1,len(intervals)):
                if intervals[i][0] < temp[1]:
                    temp[1] = max(temp[1], intervals[i][1])
                else:
                    ans.append(temp)
                    temp = intervals[i]
            ans.append(temp)
            return ans
        
        x_merge = merge_intervals(x_range)
        if len(x_merge) >= 3:
            return True
        y_merge = merge_intervals(y_range)
        if len(y_merge) >= 3:
            return True
        return False
```

##### 11. Maximum no. of overlapped Intervals at a given point of Time (Line Sweep)

```python
class Solution:
    def overlapInt(self, arr):
        events = []
        for start,end in arr:
            events.append((start,-1))
            events.append((end,1))
        
        events.sort()
        ans, temp = 0,0
        for _,x in events:
            temp -= x
            ans = max(ans,temp)
        return ans
```

##### 12. Count Days Without Meetings

```python
class Solution:
    def countDays(self, days: int, intervals: List[List[int]]) -> int:
        intervals.sort()
        n = len(intervals)
        temp = intervals[0]
        ans = 0
        for i in range(1,n):
            if temp[1] >= intervals[i][0]:
                # overlap, so merge
                temp[1] = max(temp[1],intervals[i][1])
            else:
                ans += (temp[1] - temp[0] + 1)
                temp = intervals[i]
        ans += (temp[1] - temp[0] + 1)
        return days - ans
```

### Hard

##### 1. Meeting Rooms III

```python
from heapq import heapify, heappush, heappop
class Solution:
    def mostBooked(self, n: int, meetings: List[List[int]]) -> int:
        meetings.sort()
        unused_heap = [i for i in range(n)]
        heapify(unused_heap)
        used_count = [0] * n

        min_heap = []

        for start, end in meetings:
            # free the meeting rooms
            while min_heap and min_heap[0][0] <= start:
                time, room = heappop(min_heap)
                heappush(unused_heap, room)

            if unused_heap:
                # rooms are avaiable
                room = heappop(unused_heap)
                used_count[room] += 1
                heappush(min_heap, (end,room))
            else:
                # rooms not available
                time, room = heappop(min_heap)
                heappush(min_heap, (time + end - start, room))
                used_count[room] += 1
        
        maxi, ans = -1, -1
        for i in range(n):
            if used_count[i] > maxi:
                maxi = used_count[i]
                ans = i
        return ans
```

##### 2. Minimum Interval to Include Each Query


```python
from heapq import heappush, heappop
class Solution:
    def minInterval(self, intervals: List[List[int]], queries: List[int]) -> List[int]:
        intervals.sort()
        sorted_queries = sorted((q,i) for i,q in enumerate(queries))
        ans = [-1] * len(queries)
        min_heap = []
        i,n = 0, len(intervals)

        for q,idx in sorted_queries:
            while i<n and intervals[i][0] <= q:
                start,end = intervals[i]
                heappush(min_heap,(end-start+1,end))
                i += 1            
            while min_heap and min_heap[0][1] < q:
                heappop(min_heap)
            if min_heap:
                ans[idx] = min_heap[0][0]
        return ans
```

##### 4. Employee Free Time (Famous Question - Line Sweep)

```python
"""
# Definition for an Interval.
class Interval:
    def __init__(self, start: int = None, end: int = None):
        self.start = start
        self.end = end
"""
class Solution:
    def employeeFreeTime(self, schedule: '[[Interval]]') -> '[Interval]':
        # line sweep
        events = []
        for i in range(len(schedule)):
            for j in range(len(schedule[i])):
                s,e = schedule[i][j].start, schedule[i][j].end
                events.append((s,1))
                events.append((e,-1))
        events.sort()
        ans = []
        s, prev = 0, None
        for time, x in events:
            if s == 0 and prev is not None and prev != time:
                ans.append(Interval(prev, time))
            s += x
            prev = time

        return ans
```
