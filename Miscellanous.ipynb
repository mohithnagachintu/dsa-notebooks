##### 1. Longest Palindromic String - (Expand From Center)

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        n = len(s)

        # Expand from Center
        ans, max_len = "", 0
        for i in range(n):
            l, r = i, i
            
            while l >= 0 and r < n and s[l] == s[r]:
                if r - l + 1 > max_len:
                    ans = s[l:r+1]
                    max_len = r - l + 1
                l -= 1
                r += 1
            
            l, r = i, i + 1
            while l >= 0 and r < n and s[l] == s[r]:
                if r - l + 1 > max_len:
                    ans = s[l:r+1]
                    max_len = r - l + 1
                l -= 1
                r += 1
        return ans
```

##### 2. Palindromic Substrings

```python
class Solution:
    def countSubstrings(self, s: str) -> int:
        n = len(s)
        ans = 0
        for i in range(n):
            l,r = i,i
            while l >= 0 and r < n and s[l] == s[r]:
                ans += 1
                l -= 1
                r += 1

            l,r = i,i+1
            while l >=0 and r < n and s[l] == s[r]:
                ans += 1
                l -= 1
                r += 1
        return ans
```
