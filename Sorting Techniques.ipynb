##### Line Sweep

##### Restaurant Customers - https://cses.fi/problemset/task/1619

```python
def max_customers(n, customers):
    events = []

    for a,b in customers:
        events.append((a,1))   # arrival
        events.append((b,-1))  # leaves
    
    # Sorting the events: first by time, then by type (exit before entry in case of tie)
    events.sort()
    # print(events)

    # sweep line algorithm
    maxi,temp = 0,0
    for x,event in events:
        temp += event
        maxi = max(maxi, temp)
        # print(x,event, temp, maxi)
    return maxi

# 3
# 5 8
# 2 4
# 3 9
n = int(input())
customers = [tuple(map(int, input().split())) for _ in range(n)]
print(max_customers(n, customers))
```
