```python
class DisjointSet:
    def __init__(self):
        self.parent = dict()
        self.rank = dict()
        self.size = dict()
    
    def find(self,node):
        if node not in self.parent:
            self.parent[node] = node
            self.rank[node] = 0
            self.size[node] = 1

        if node != self.parent[node]:
            self.parent[node] = self.find(self.parent[node])
        return self.parent[node]
    
    def union(self, u, v, by_size = True):
        root_u = self.find(u)
        root_v = self.find(v)
        if root_u == root_v:
            return

        if by_size:
            if self.size[root_u] < self.size[root_v]:
                self.parent[root_u] = root_v
                self.size[root_v] += self.size[root_u]
            else:
                self.parent[root_v] = root_u
                self.size[root_u] += self.size[root_v]
        else:
            if self.rank[root_u] < self.rank[root_v]:
                self.parent[root_u] = root_v
            elif self.rank[root_u] > self.rank[root_v]:
                self.parent[root_v] = root_u
            else:
                self.parent[root_v] = root_u
                self.rank[root_u] += 1 

    def connected(self, u, v):
        return self.find(u) == self.find(v)


```

```python
def count_connected_components(nodes, edges):
    ds = DisjointSet()
    
    # 1. Initialize all nodes in the Disjoint Set
    # find() handles initialization in your provided class
    for node in nodes:
        ds.find(node)
    
    # Track unique components. Initially, each node is its own component.
    num_components = len(nodes)
    
    # 2. Process each edge
    for u, v in edges:
        root_u = ds.find(u)
        root_v = ds.find(v)
        
        # If they are in different sets, union them and reduce the count
        if root_u != root_v:
            ds.union(u, v)
            num_components -= 1
            
    return num_components

# --- Example Usage ---
nodes_list = [0, 1, 2, 3, 4]
edges_list = [(0, 1), (1, 2), (3, 4)]

result = count_connected_components(nodes_list, edges_list)
print(f"Number of connected components: {result}") # Output: 2
```
