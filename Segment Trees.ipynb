```python
MAX = float('inf')

def construct_segment_tree(arr):
    n = len(arr)
    if n == 0: return []
    if (n & (n-1) == 0):
        size = n
    else:
        size = 1 << (n-1).bit_length()
    seg_size = 2 * size - 1
    seg = [MAX] * seg_size
    
    def helper(low,high,i):
        if low == high:
            seg[i] = arr[low]
            return
        mid = (low + high) // 2
        left_child = 2 * i + 1
        right_child = 2 * i + 2
        helper(low, mid, left_child)
        helper(mid+1, high, right_child)
        seg[i] = min(seg[left_child], seg[right_child])

    helper(0,n-1,0)
    return seg

```

```python
construct_segment_tree([7, 2, 9, 4, 6])
```
