##### 1. Binary Search

```python
arr = [10,20,30,40,50]

def binary_search(arr,target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low+high)//2
        if target == arr[mid]: 
            return mid
        elif target < arr[mid]: 
            high = mid - 1
        else: 
            low = mid + 1
    return -1

# Testing
for x in [20,60,10]:
    print(binary_search(arr,x))
```

###### 2. Implement Lower bound

```python
# Lower bound: smallest index idx where arr[idx] >= num
def lower_bound(arr,num):
    low,high,ans = 0,len(arr)-1,len(arr)
    while low <= high: 
        mid = (low + high)//2
        if (arr[mid] >= num):
            ans = mid
            high = mid - 1
        else: 
            low = mid + 1
    return ans

arr = [6, 12]
lower_bound(arr,12)
```

##### 3. Implement Upper Bound

```python
# Upper Bound: smallest index idx where arr[idx] > num
def upper_bound(arr,num):
    low, high,ans = 0, len(arr) - 1,-1
    while low <= high:
        mid = (low+high)//2
        if (arr[mid] > num):
            ans = mid
            high = mid - 1
        else: 
            low = mid + 1
    return ans
arr = [1,2,3,4,5,6,7]
upper_bound(arr,5)
```

##### 6. First / last occurrence of a number

```python
def searchRange(nums, target):
    def first_occurrence(nums,target):
        low,high,ans = 0,len(nums)-1,-1
        while low <= high:
            mid = (low+high)//2
            if (nums[mid]==target):
                ans = mid
                high = mid - 1
            elif (nums[mid]>target):
                high = mid - 1
            else:
                low = mid + 1
        return ans
    def last_occurrence(nums,target):
        low,high,ans = 0,len(nums)-1,-1
        while low<=high:
            mid = (low+high)//2
            if (nums[mid]==target):
                ans = mid
                low = mid + 1
            elif (nums[mid]>target):
                high = mid - 1
            else:
                low = mid + 1
        return ans

    return [first_occurrence(nums,target),last_occurrence(nums,target)]
```

##### 8. Search in Rotated Sorted Array - I

```python
# key point is here that, if array is rotated atleast one half is rotated.
def search(nums, target):
    low, high = 0, len(nums) - 1
    while low <= high:
        mid = (low + high)//2
        if nums[mid] == target:
            return mid
        if nums[low] <= nums[mid]:
            # left half is sorted search in left half
            if target >= nums[low] and target < nums[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            # right half is sorted, search in right half
            if target > nums[mid] and target <= nums[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1
```

##### 12. Single Element in a Sorted Array

```python
# You are given a sorted array consisting of only integers where every element appears exactly twice, 
# except for one element which appears exactly once.Return the single element that appears only once.
# Your solution must run in O(log n) time and O(1) space.

def singleNonDuplicate(nums):
    # [1,1,2,3,3,4,4,8,8]
    #  E,O,E,O,E,O,E,O,E
    # (Even,Odd) <---(break in pattern)--->(Odd,Even)
    # if nums[mid] not equals to any of adjacent it's the answer
    # if two numbers index is (even,odd), then target is on right half 
    # if two numbers index is (odd,even), then target is on left half

    n = len(nums)
    #  Edge cases
    if n==1: return nums[0]
    if nums[0] != nums[1]: return nums[0]
    if nums[n-1] != nums[n-2]: return nums[n-1]
    low,high = 1, n - 2
    while (low <= high):
        mid = (low + high)//2
        if nums[mid] != nums[mid+1] and nums[mid] != nums[mid-1]:
            return nums[mid]
        # conditions to satsify for right half
        if (mid & 1 == 0 and nums[mid] == nums[mid+1]) or (mid & 1 and nums[mid]==nums[mid-1]):
            low = mid + 1
        else:
            high = mid -1
    return -1
```

##### 13. Find Peak Element

```python
def findPeakElement(nums) -> int:
    # -inf  , [nums] , -inf

    n = len(nums)
    if n==1: return 0
    if nums[0] > nums[1]: return 0
    if nums[n-1] > nums[n-2]: return n-1

    # since we are comparing mid with mid-1 and mid + 1, 
    # we should start with 1, n-2 to avoid overflow

    low,high = 1,n-2
    while low <= high:
        mid = (low + high)//2
        if (nums[mid] > nums[mid-1] and nums[mid] > nums[mid+1]):
            return mid
        elif nums[mid-1] < nums[mid]:
            low = mid + 1
        else:
            high = mid - 1

    return -1
```

##### 16. Koko Eating Bananas

```python
class Solution:
    def helper(self,piles,x,h):
        count = 0
        for pile in piles:
            count += (pile//x)
            if pile % x != 0:
                count += 1
        return count <= h    
    def minEatingSpeed(self, piles, h):
        piles.sort()
        low, high,ans = 1, piles[-1],piles[-1]
        while low <= high:
            mid = (low+high)//2
            if self.helper(piles,mid,h):
                ans = mid
                high = mid - 1
            else:
                low = mid + 1
        return ans
```

```python
# Leetcode - 1552. Magnetic Force Between Two Balls

def maxDistance(position,m):
    position.sort()
    
    def isPossible(position,gap,m):
        cnt, last_placed = 1, position[0]
        for i in range(1,len(position)):
            if position[i] - last_placed >= gap:
                last_placed = position[i]
                cnt += 1
            else:  # skip this position
                pass
            if cnt >= m:
                return True
        return False

    # dist -- as a binary search
    low, high  = 1, position[-1]-position[0]
    ans = high
    while low <= high:
        mid = (low + high)//2
        print(low,mid,high,ans)
        if (isPossible(position,mid,m)):
            ans = mid
            low = mid + 1
        else:
            high = mid - 1

    return ans
```

##### 23. Split array - Largest Sum

```python
def splitArray(nums,k):
    def splitArrayHelper(maxi):
        sub_arrays, sub_array_sum = 1, 0
        for num in nums:
            if sub_array_sum + num <= maxi:
                sub_array_sum += num
            else:
                sub_arrays += 1
                sub_array_sum = num
        return sub_arrays

    # trail and error method 
    low, high = max(nums), sum(nums)
    # mid -> maximum sum of the subarray
    # split the array with sum of subarray with max cap of mid
    while low <= high:
        mid = (low + high)//2
        if splitArrayHelper(mid) > k:
            low = mid + 1
        else:
            high = mid - 1
    return low
    # T(N) = O(N*log(K)), where K = sum(nums) - max(nums) 
```

##### 29. Search in a sorted 2D matrix

```python
def searchMatrix(mat, target):
    # flatten the 2d matrix
    m,n = len(mat),len(mat[0])
    low, high = 0, n*m-1
    while low <= high:
        mid = (low + high) // 2
        row = mid // n
        col = mid % n
        if target == mat[row][col]:
            return True
        elif target > mat[row][col]:
            low = mid + 1
        else:
            high = mid - 1
    return False
```

##### 30. Search in a row and column-wise sorted matrix

```python
def searchMatrix(matrix, target):
    m,n = len(matrix),len(matrix[0])
    i,j = 0,n-1
    while i<m and j>=0:
        if target < matrix[i][j]:
            j -= 1
        elif target > matrix[i][j]:
            i += 1
        else:
            return True
    return False
```

##### 31. Find Peak Element (2D Matrix)

```python
def findPeakGrid(mat):
    # worst - m * n, average - m * logn
    def col_maximum(j):
        maxi = -1
        idx = -1
        for i in range(len(mat)):
            if maxi < mat[i][j]:
                maxi,idx = mat[i][j],i
        return idx

    m, n = len(mat), len(mat[0])
    low,high = 0, n - 1
    while low <= high:
        mid = (low + high)//2
        idx = col_maximum(mid)
        maxi = mat[idx][mid]
        left, right = -1, -1
        if mid-1 >= 0:
            left = mat[idx][mid-1]
        if mid + 1 < n:
            right = mat[idx][mid+1]
        if maxi > left and maxi > right:
            return [idx,mid]
        elif maxi < left:
            high = mid - 1
        else:
            low = mid + 1
    return [-1,-1]
```
