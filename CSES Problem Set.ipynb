# 1. Sorting and Searching

##### 1. Distinct Numbers

```python
import sys

input = sys.stdin.buffer.readline
write = sys.stdout.write

n = int(input())
a = list(map(int, input().split()))

a.sort()

distinct = 1 if n else 0
for i in range(1, n):
    if a[i] != a[i - 1]:
        distinct += 1

write(f"{distinct}\n")
```
