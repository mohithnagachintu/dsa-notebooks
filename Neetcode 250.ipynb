##### 1. Concatenation of Array

```python
class Solution:
    def getConcatenation(self, nums: List[int]) -> List[int]:
        n = len(nums)
        ans = [None] * (n*2)
        for i in range(n):
            ans[i] = nums[i]
            ans[i+n] = nums[i]
        return ans
```

##### 2. Contains Duplicate

```python
class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        st = set()
        for x in nums:
            if x in st:
                return True
            st.add(x)
        return False
```

##### 3. Valid Anagram

```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        n,m = len(s), len(t)

        if n != m:
            return False

        mapper = defaultdict(int)
        for i in range(n):
            mapper[s[i]] += 1
        for i in range(n):
            mapper[t[i]] -= 1
            if mapper[t[i]] == 0:
                del mapper[t[i]]
        return True if len(mapper) == 0 else False
    
```

##### 4. Two Sum

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        hash_map = dict()
        for i in range(len(nums)):
            if target - nums[i] in hash_map:
                return [i, hash_map[target-nums[i]]]
            hash_map[nums[i]] = i
        return [-1,-1]
```

##### 5. Longest Common Prefix

```python
class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        n = len(strs)

        for i,ch in enumerate(strs[0]):
            for j in range(1,n):
                s = strs[j]
                if i >= len(s) or s[i] != ch:
                    return strs[0][:i]
        return strs[0]
```

##### 7. Remove Element

```python
class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:
        i,j = 0,len(nums)-1
        while i <= j:
            if nums[i] == val:
                # swap nums[i] , nums[j]
                nums[i],nums[j] = nums[j],nums[i]
                j -= 1
            else:
                i += 1
        return j + 1
```

##### 8. Majority element

```python
class Solution:
    # Boyer-Moore Voting Algorithm
    def majorityElement(self, nums: List[int]) -> int:
        count = 0
        ans = None

        for num in nums:
            if count == 0:
                ans = num
            if num == ans:
                count += 1
            else:
                count -= 1
        return ans
```

##### 9. Majority Element 2

```python
class Solution:
    def majorityElement(self, nums: List[int]) -> List[int]:
        # for a array length of n, 
        # There can be at most one majority element which is more than ⌊n/2⌋ times.
        # There can be at most two majority elements which are more than ⌊n/3⌋ times.
        # There can be at most three majority elements which are more than ⌊n/4⌋ times

        count1, count2 = 0,0
        ans1, ans2 = None, None

        for num in nums:

            if num == ans1:
                count1 += 1
            elif num == ans2:
                count2 += 1
            elif count1 == 0:
                count1 += 1
                ans1 = num
            elif count2 == 0:
                count2 += 1
                ans2 = num
            else:
                count1 -= 1
                count2 -= 1

        res = []
        for ans in [ans1,ans2]:
            if nums.count(ans) > len(nums) // 3:
                res.append(ans)
        return res
```

##### 11. Encode and Decode Strings

```python
class Codec:
    def encode(self, strs: List[str]) -> str:
        """Encodes a list of strings to a single string.
        """
        ans = []
        for s in strs:
            ans.append(str(len(s)))
            ans.append('@')
            ans.append(s)
        return ''.join(ans)

    def decode(self, s: str) -> List[str]:
        """Decodes a single string to a list of strings.
        """
        ans = []
        i,n = 0,len(s)
        while i < n:
            count_str = []
            while s[i] != '@':
                count_str.append(s[i])
                i += 1
            i += 1 # this is @
            count = int(''.join(count_str))
            ans.append(s[i:i+count])
            i += count
        return ans


# Your Codec object will be instantiated and called as such:
# codec = Codec()
# codec.decode(codec.encode(strs))
```

##### 12. Range Sum Query 2D - Immutable

```python
class NumMatrix:

    def __init__(self, matrix: List[List[int]]):
        m,n = len(matrix), len(matrix[0])
        self.prefix_2d = [[0 for _ in range(n+1)] for _ in range(m+1)]
        for i in range(m):
            p = 0
            for j in range(n):
                p += matrix[i][j]
                self.prefix_2d[i+1][j+1] = p + self.prefix_2d[i][j+1]

    def sumRegion(self, r1: int, c1: int, r2: int, c2: int) -> int:
        bottom_right = self.prefix_2d[r2+1][c2+1]
        bottom_left = self.prefix_2d[r2+1][c1]
        top_right = self.prefix_2d[r1][c2+1]
        top_left = self.prefix_2d[r1][c1]
        return bottom_right - bottom_left - top_right + top_left
```

##### 15. Longest Consecutive Sequence

```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        nums_set = set(nums)
        ans = 0

        for num in nums_set:
            if num-1 not in nums_set:
                start = 1
                curr = num

                while curr + 1 in nums_set:
                    curr += 1
                    start += 1
                ans = max(ans, start)
        return ans
```

##### 1. Reverse a string

```python
class Solution:
    def reverseString(self, s: List[str]) -> None:
        """
        Do not return anything, modify s in-place instead.
        """
        left, right = 0,len(s)-1
        while left < right:
            s[left],s[right] = s[right],s[left]
            left += 1
            right -= 1
        
```

##### 2. Valid Palindrome

```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        left, right = 0, len(s) - 1
        while left < right:
            while left < right and s[left].isalnum() == False:
                left += 1
            while left < right and  s[right].isalnum() == False:
                right -= 1
            if s[left].lower() != s[right].lower():
                return False
            else:
                left += 1
                right -= 1
        return True
```

##### 3. Valid Palindrome 2

```python
class Solution:
    def validPalindrome(self, s: str) -> bool:
        
        def is_palindrome(s,left,right):
            while left < right and s[left] == s[right]:
                left += 1
                right -= 1
            return True if left >= right else False

        left, right = 0, len(s) - 1
        while left < right:
            if s[left] == s[right]:
                left += 1
                right -= 1
            else:
                skip_left = is_palindrome(s,left+1,right)
                skip_right = is_palindrome(s,left,right-1)
                return skip_left or skip_right
        return True
```

##### 4. Merge Strings Alternately

```python
class Solution:
    def mergeAlternately(self, word1: str, word2: str) -> str:
        ans = []
        m,n = len(word1),len(word2)
        i = 0
        while i < min(m,n):
            ans.append(word1[i])
            ans.append(word2[i])
            i += 1
        j = i
        while i < m:
            ans.append(word1[i])
            i += 1
        while j < n:
            ans.append(word2[j])
            j += 1
        return ''.join(ans)
```

##### 5. Merge Sorted Array

```python
class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        # TC - O(m+n), SC - O(1)
        # 3 pointers - end
        p1, p2 = m-1,n-1
        for p3 in range(m+n-1,-1,-1):
            if p2 < 0:
                break
            if p1 >= 0 and nums1[p1] > nums2[p2]:
                nums1[p3] = nums1[p1]
                p1 -= 1
            else:
                nums1[p3] = nums2[p2]
                p2 -= 1
```

##### 6. Remove Duplicates from a sorted array

```python
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        # TC - O(N) , SC - O(1)
        n = len(nums)
        next_unique_idx = 1
        for i in range(1,n):
            # move unique element to the top
            if nums[i-1] != nums[i]:
                nums[next_unique_idx] = nums[i]
                next_unique_idx += 1
        return next_unique_idx
```

##### 1. Longest Common Prefix (Trie Sol - But overkill)

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()
    
    def insert(self, word):
        cur = self.root
        for ch in word:
            if ch not in cur.children:
                cur.children[ch] = TrieNode()
            cur = cur.children[ch]
        cur.end = True
    
class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        trie = Trie()
        for word in strs:
            if len(word) == 0:
                return ""
            trie.insert(word)

        cur = trie.root
        ans = []
        # edge case - ["ab","a"]
        while len(cur.children) == 1 and not cur.end:
            for k,v in cur.children.items():
                ans.append(k)
                cur = cur.children[k]
        return ''.join(ans)
```
