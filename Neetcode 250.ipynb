##### 1. Concatenation of Array

```python
class Solution:
    def getConcatenation(self, nums: List[int]) -> List[int]:
        n = len(nums)
        ans = [None] * (n*2)
        for i in range(n):
            ans[i] = nums[i]
            ans[i+n] = nums[i]
        return ans
```

##### 2. Contains Duplicate

```python
class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        st = set()
        for x in nums:
            if x in st:
                return True
            st.add(x)
        return False
```

##### 3. Valid Anagram

```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        n,m = len(s), len(t)

        if n != m:
            return False

        mapper = defaultdict(int)
        for i in range(n):
            mapper[s[i]] += 1
        for i in range(n):
            mapper[t[i]] -= 1
            if mapper[t[i]] == 0:
                del mapper[t[i]]
        return True if len(mapper) == 0 else False
    
```

##### 4. Two Sum

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        hash_map = dict()
        for i in range(len(nums)):
            if target - nums[i] in hash_map:
                return [i, hash_map[target-nums[i]]]
            hash_map[nums[i]] = i
        return [-1,-1]
```

##### 5. Longest Common Prefix

```python
class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        n = len(strs)

        for i,ch in enumerate(strs[0]):
            for j in range(1,n):
                s = strs[j]
                if i >= len(s) or s[i] != ch:
                    return strs[0][:i]
        return strs[0]
```

##### 7. Remove Element

```python
class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:
        i,j = 0,len(nums)-1
        while i <= j:
            if nums[i] == val:
                # swap nums[i] , nums[j]
                nums[i],nums[j] = nums[j],nums[i]
                j -= 1
            else:
                i += 1
        return j + 1
```

##### 8. Majority element

```python
class Solution:
    # Boyer-Moore Voting Algorithm
    def majorityElement(self, nums: List[int]) -> int:
        count = 0
        ans = None

        for num in nums:
            if count == 0:
                ans = num
            if num == ans:
                count += 1
            else:
                count -= 1
        return ans
```

##### 9. Majority Element 2

```python
class Solution:
    def majorityElement(self, nums: List[int]) -> List[int]:
        # for a array length of n, 
        # There can be at most one majority element which is more than ⌊n/2⌋ times.
        # There can be at most two majority elements which are more than ⌊n/3⌋ times.
        # There can be at most three majority elements which are more than ⌊n/4⌋ times

        count1, count2 = 0,0
        ans1, ans2 = None, None

        for num in nums:

            if num == ans1:
                count1 += 1
            elif num == ans2:
                count2 += 1
            elif count1 == 0:
                count1 += 1
                ans1 = num
            elif count2 == 0:
                count2 += 1
                ans2 = num
            else:
                count1 -= 1
                count2 -= 1

        res = []
        for ans in [ans1,ans2]:
            if nums.count(ans) > len(nums) // 3:
                res.append(ans)
        return res
```

##### 12. Range Sum Query 2D - Immutable

```python
class NumMatrix:

    def __init__(self, matrix: List[List[int]]):
        m,n = len(matrix), len(matrix[0])
        self.prefix_2d = [[0 for _ in range(n+1)] for _ in range(m+1)]
        for i in range(m):
            p = 0
            for j in range(n):
                p += matrix[i][j]
                self.prefix_2d[i+1][j+1] = p + self.prefix_2d[i][j+1]

    def sumRegion(self, r1: int, c1: int, r2: int, c2: int) -> int:
        bottom_right = self.prefix_2d[r2+1][c2+1]
        bottom_left = self.prefix_2d[r2+1][c1]
        top_right = self.prefix_2d[r1][c2+1]
        top_left = self.prefix_2d[r1][c1]
        return bottom_right - bottom_left - top_right + top_left
```

##### 15. Longest Consecutive Sequence

```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        nums_set = set(nums)
        ans = 0

        for num in nums_set:
            if num-1 not in nums_set:
                start = 1
                curr = num

                while curr + 1 in nums_set:
                    curr += 1
                    start += 1
                ans = max(ans, start)
        return ans
```

##### 2. Valid Palindrome

```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        left, right = 0, len(s) - 1
        while left < right:
            while left < right and s[left].isalnum() == False:
                left += 1
            while left < right and  s[right].isalnum() == False:
                right -= 1
            if s[left].lower() != s[right].lower():
                return False
            else:
                left += 1
                right -= 1
        return True
```

##### 3. Valid Palindrome 2

```python
class Solution:
    def validPalindrome(self, s: str) -> bool:
        
        def is_palindrome(s,left,right):
            while left < right and s[left] == s[right]:
                left += 1
                right -= 1
            return True if left >= right else False

        left, right = 0, len(s) - 1
        while left < right:
            if s[left] == s[right]:
                left += 1
                right -= 1
            else:
                skip_left = is_palindrome(s,left+1,right)
                skip_right = is_palindrome(s,left,right-1)
                return skip_left or skip_right
        return True
```

##### 5. Merge Sorted Array

```python
class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        # TC - O(m+n), SC - O(1)
        # 3 pointers - end
        p1, p2 = m-1,n-1
        for p3 in range(m+n-1,-1,-1):
            if p2 < 0:
                break
            if p1 >= 0 and nums1[p1] > nums2[p2]:
                nums1[p3] = nums1[p1]
                p1 -= 1
            else:
                nums1[p3] = nums2[p2]
                p2 -= 1
```

##### 6. Remove Duplicates from a sorted array

```python
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        # TC - O(N) , SC - O(1)
        n = len(nums)
        next_unique_idx = 1
        for i in range(1,n):
            # move unique element to the top
            if nums[i-1] != nums[i]:
                nums[next_unique_idx] = nums[i]
                next_unique_idx += 1
        return next_unique_idx
```

##### 8. Three Sum

```python
class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        nums.sort()
        n = len(nums)
        ans = []
        target = 0
        i = 0
        while i < n:
            left, right = i+1, n - 1
            while left < right:
                s = nums[left] + nums[right] + nums[i]
                if s < target:
                    left += 1
                elif s > target:
                    right -= 1
                else:
                    ans.append([nums[i],nums[left],nums[right]])

                    # remove duplicates of low and high
                    while left < right and ans[-1][1] == nums[left]:
                        left += 1
                    while left < right and ans[-1][2] == nums[right]:
                        right -= 1
            # remove duplicates of i
            while i + 1 < n and nums[i] == nums[i+1]:
                i += 1
            i += 1
        return ans
```

##### 9. Four Sum

```python
class Solution:
    def fourSum(self, nums: List[int], target: int) -> List[List[int]]:
        nums.sort()
        n = len(nums)
        ans = []
        i = 0
        while i < n:
            j = i + 1
            while j < n:
                left, right = j + 1, n - 1
                while left < right:
                    s = nums[i] + nums[j] + nums[left] + nums[right]
                    if s < target:
                        left += 1
                    elif s > target:
                        right -= 1
                    else:
                        ans.append([
                            nums[i],
                            nums[j],
                            nums[left],
                            nums[right]
                        ])
                        # remove duplicates of left and right
                        while left < right and nums[left] == ans[-1][2]:
                            left += 1
                        while left < right and nums[right] == ans[-1][3]:
                            right -= 1
                # remove duplicates of j
                while j + 1 < n and nums[j] == nums[j + 1]:
                    j += 1
                j += 1
            # remove duplicates of i
            while i + 1 < n and nums[i] == nums[i + 1]:
                i += 1
            i += 1
        return ans
```

##### 10. Rotate 1D Array - Important

```python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        def reverse(nums,left,right):
            while left < right:
                nums[left],nums[right] = nums[right], nums[left]
                left += 1
                right -= 1
        n = len(nums)
        k = k % n
        reverse(nums,0,n-1)
        reverse(nums,0,k-1)
        reverse(nums,k,n-1)
        # TC - O(N+N) ~ O(N)
```

##### 11. Container With Most Water

```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        left, right = 0, len(height) - 1
        ans = 0
        while left < right:
            temp = min(height[left], height[right]) * (right - left) 
            ans = max(ans, temp)

            if height[left] < height[right]:
                left += 1
            else:
                right -= 1
        return ans
```

##### 13. Trapping Rain Water - Very Very Important

```python
class Solution:
    def trap(self, height: List[int]) -> int:
        n = len(height)
        left, right = [0] * n, [0] * n
        left[0], right[-1] = height[0], height[-1]

        for i in range(1,n):
            left[i] = max(left[i-1], height[i])
        for j in range(n-2,-1,-1):
            right[j] = max(right[j+1],height[j])
        ans = 0
        # print(left)
        # print(right)
        for k in range(1,n-1):
            water_trapped = min(left[k],right[k]) - height[k]
            # print(f'Water Trapped at {k} - {water_trapped}')
            ans += water_trapped
        return ans
```

# 3. Sliding Window

##### 1. Contains Duplicates

```python
class Solution:
    def containsNearbyDuplicate(self, nums: List[int], k: int) -> bool:
        window_set = set()
        i,j = 0,0
        while i <= j and j < len(nums):
            if j-i > k:
                window_set.remove(nums[i])
                i += 1
            if nums[j] in window_set:
                return True
            window_set.add(nums[j])
            j += 1
        return False
```

##### 2. Longest Substring without Repeating Characters

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        n = len(s)
        i,j = 0,0
        ans = 0
        window_set = set()
        while i <= j and j < n:
            while s[j] in window_set:
                window_set.remove(s[i])
                i += 1
            window_set.add(s[j])
            ans = max(ans,j-i+1)
            j += 1
        return ans
```

##### 3. Longest Repeating Character Replacement

```python
class Solution:
    def characterReplacement(self, s: str, k: int) -> int:
        freq = defaultdict(int)
        n = len(s)
        i = j = maxi = ans = 0
        while i <= j and j < n:
            freq[s[j]] += 1
            maxi = max(maxi, freq[s[j]])
            # no. of replacements > k, shrink the window
            if j - i + 1 - maxi > k:
                freq[s[i]] -= 1
                i += 1
            ans = max(ans,j-i+1)
            j += 1
        return ans
```

##### 5. Minimal Size Subarray Sum

```python
class Solution:
    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        ans = float('inf')
        left = 0
        s = 0
        for right in range(len(nums)):
            s += nums[right] 
            while s >= target:
                s -= nums[left]
                ans = min(ans, right - left + 1)
                left += 1
        if ans > 1e9:
            return 0
        else:
            return ans
```

##### 5. Sliding Window Maximum

```python
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        # monotonic decreasing queue
        dq = deque()
        ans = []
        for i in range(k):
            while dq and nums[i] >= nums[dq[-1]]:
                dq.pop()
            dq.append(i)
        ans.append(nums[dq[0]])

        for i in range(k,len(nums)):
            if dq and dq[0] == i - k:
                dq.popleft()
            while dq and nums[i] >= nums[dq[-1]]:
                dq.pop()
            dq.append(i)
            ans.append(nums[dq[0]])
        return ans
```

##### 6. Sum of Subarray minimums

```python
MOD = 1e9 + 7
class Solution:
    def sumSubarrayMins(self, arr: List[int]) -> int:
        # idea: in how many subarrays arr[i] is the minimum

        n = len(arr)
        nse,pse = [n] * n, [-1] * n
        st = [0]
        i = 1
        # previous smaller element
        while i < n:
            while st and arr[st[-1]] >= arr[i]:
                st.pop()
            if st:
                pse[i] = st[-1]
            st.append(i)
            i += 1
        i = 1
        st = [0]
        # next smaller or equal element
        while i < n:
            while st and arr[st[-1]] >= arr[i]:
                nse[st[-1]] = i
                st.pop()
            st.append(i)
            i += 1
        ans = 0
        for i in range(n):
            left = i - pse[i]
            right = nse[i] - i
            subarrays = left * right
            ans = (ans + arr[i] * subarrays) % MOD
        return int(ans)
```

##### 7. Largest Area in a Rectangle

```python
class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        # Each bar tries to expand left & right until smaller bar blocks
        # width = right[i] - left[i] - 1
        # area = height[i] * width
        # left = previous smaller element of i
        # right = next smaller element of i
        n = len(heights)
        nse, pse = [n] * n, [-1] * n
        st = [0]
        i = 1
        while i < n:
            while st and heights[st[-1]] > heights[i]:
                nse[st[-1]] = i
                st.pop()
            st.append(i)
            i += 1
        st.clear()
        st.append(0)
        i = 1
        while i < n:
            while st and heights[st[-1]] >= heights[i]:
                st.pop()
            if st:
                pse[i] = st[-1]
            st.append(i)
            i += 1
        ans = 0
        for i in range(n):
            width = nse[i] - pse[i] - 1
            area = heights[i] * width
            ans = max(area, ans)
        return ans
```

##### 1. Stack using Queues

```python
class Queue:
    def __init__(self):
        self.queue = deque()
    
    def enqueue(self,x):
        self.queue.append(x)

    def dequeue(self):
        return self.queue.popleft()
    
    def __len__(self):
        return len(self.queue)

class MyStack:
    def __init__(self):
        self.queue = Queue()

    def push(self, x: int) -> None:
        self.queue.enqueue(x)

    def pop(self) -> int:
        n = len(self.queue)
        for _ in range(n-1):
            x = self.queue.dequeue()
            self.queue.enqueue(x)
        return self.queue.dequeue()
        
    def top(self) -> int:
        n = len(self.queue)
        for _ in range(n-1):
            x = self.queue.dequeue()
            self.queue.enqueue(x)
        top = self.queue.dequeue()
        self.queue.enqueue(top)
        return top

    def empty(self) -> bool:
        return len(self.queue) == 0
```

##### 2. Implement Queue using Stack

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, x):
        self.stack.append(x)

    def pop(self):
        return self.stack.pop()

    def top(self):
        return self.stack[-1]

    def __len__(self):
        return len(self.stack)


class MyQueue:

    def __init__(self):
        self.in_stack = Stack()
        self.out_stack = Stack()

    def _move(self):
        if len(self.out_stack) == 0:
            while len(self.in_stack):
                self.out_stack.push(self.in_stack.pop())

    def push(self, x: int) -> None:
        self.in_stack.push(x)

    def pop(self) -> int:
        self._move()
        return self.out_stack.pop()

    def peek(self) -> int:
        self._move()
        return self.out_stack.top()

    def empty(self) -> bool:
        return len(self.in_stack) == 0 and len(self.out_stack) == 0
```

##### 4. Evaluate Reverse Polish Notation

```python
class Solution:
    def evalRPN(self, tokens: List[str]) -> int:
        n = len(tokens)
        stack = []
        ops = {
            '+': lambda a, b: a + b,
            '-': lambda a, b: a - b,
            '*': lambda a, b: a * b,
            '/': lambda a, b: int(a / b)
        }
        for t in tokens:
            if t in ops:
                b = stack.pop()
                a = stack.pop()
                stack.append(ops[t](a, b))
            else:
                stack.append(int(t))
        return stack[0]
```

```python
class Solution:
    def asteroidCollision(self, asteroids: List[int]) -> List[int]:
        stack = []
        for a in asteroids:
            while stack and a < 0 < stack[-1]:
                if stack[-1] < -a:
                    # top
                    stack.pop()
                    continue
                elif stack[-1] == -a:
                    # both
                    stack.pop()
                break
            else:
                stack.append(a)
        return stack
```

##### 7. Maximum Frequency Stack

```python
class FreqStack:

    def __init__(self):
        self.freq = defaultdict(int)
        self.count = defaultdict(list)
        self.maxi = 0
        

    def push(self, val: int) -> None:
        self.freq[val] += 1
        f = self.freq[val]
        if f > self.maxi:
            self.maxi = f
        self.count[f].append(val)

    def pop(self) -> int:
        popped = self.count[self.maxi].pop()
        self.freq[popped] -= 1
        if not self.count[self.maxi]:
            self.maxi -= 1
        return popped
```

##### 13. Delete a node in BST

```python
class Solution:
    def deleteNode(self, root: Optional[TreeNode], key: int) -> Optional[TreeNode]:
        if root is None:
            return None
        
        if key < root.val:
            root.left = self.deleteNode(root.left, key)
        elif key > root.val:
            root.right = self.deleteNode(root.right, key)
        else:
            if root.left is None:
                return root.right
            if root.right is None:
                return root.left
            
            # both exists, replace target with lowest value in the right sub
            successor = root.right
            while successor.left:
                successor = successor.left
            root.val = successor.val
            root.right = self.deleteNode(root.right, successor.val)

        return root
```

##### 16. Vertical Order Traversal

```python
class Solution:
    def verticalTraversal(self, root: Optional[TreeNode]) -> List[List[int]]:
        nodes = []
        def solve(root, col, row):
            if root is None:
                return
            nodes.append((col,row,root.val))
            solve(root.left, col-1, row+1)
            solve(root.right, col+1, row+1)
        solve(root,0,0)
        nodes.sort()
        ans = []
        prev_col = float('-inf')
        temp = []
        for col, row, val in nodes:
            if col != prev_col:
                if temp:
                    ans.append(temp)
                temp = []
                prev_col = col
            temp.append(val)
        if temp:
            ans.append(temp)
        return ans
```

##### 25. Serialize and Deserialize Binary Tree


```python
class Codec:

    def serialize(self, root):
        """Encodes a tree to a single string.
        
        :type root: TreeNode
        :rtype: str
        """
        st = []
        def solve(root):
            if not root:
                st.append("N")
                return
            st.append(str(root.val))
            solve(root.left)
            solve(root.right)
        solve(root)
        return ','.join(st)

    def deserialize(self, data):
        """Decodes your encoded data to tree.
        
        :type data: str
        :rtype: TreeNode
        """
        vals = data.split(',')
        i = [0] 

        def solve():
            if vals[i[0]] == 'N':
                i[0] += 1
                return None
            node = TreeNode(int(vals[i[0]]))
            i[0] += 1
            node.left = solve()
            node.right = solve()
            return node

        return solve()
```

```python
class KthLargest:

    def __init__(self, k: int, nums: List[int]):
        self.min_heap = nums
        self.k = k
        heapify(self.min_heap)
        while len(self.min_heap) > k:
            heappop(self.min_heap)        

    def add(self, val: int) -> int:
        heappush(self.min_heap, val)
        while len(self.min_heap) > self.k:
            heappop(self.min_heap)
        return self.min_heap[0]
```

##### 2. Last Stone Weight

```python
class Solution:
    def lastStoneWeight(self, stones: List[int]) -> int:
        max_heap = []
        for s in stones:
            heappush(max_heap, -s)
        while len(max_heap) > 1:
            a = -heappop(max_heap)
            b = -heappop(max_heap)
            if a != b:
                heappush(max_heap, -(a-b))
        if len(max_heap) == 0:
            return 0
        return -heappop(max_heap)
```

##### 3. K Closest Points to Origin

```python
class Solution:
    def kClosest(self, points: List[List[int]], k: int) -> List[List[int]]:
        min_heap = []
        heapify(min_heap)
        for i in range(len(points)):
            x, y = points[i]
            heappush(min_heap, (math.sqrt(x**2 + y**2), i))
        ans = []
        for i in range(k):
            _, idx = heappop(min_heap)
            ans.append(points[idx])
        return ans
```

##### 5. Task Scheduler

```python
class Solution:
    def leastInterval(self, tasks: List[str], n: int) -> int:
        counts = Counter(tasks)
        max_heap = [ -cnt for cnt in counts.values()]
        heapify(max_heap)
        # (cnt, next_available_time)
        cool_down_queue = deque()
        time = 0
        while max_heap or cool_down_queue:
            time += 1
            if max_heap:
                count = -heappop(max_heap)-1
                if count:
                    cool_down_queue.append((-count,time+n))
            if cool_down_queue and cool_down_queue[0][1] == time:
                heappush(max_heap,cool_down_queue.popleft()[0])
        return time
```

##### 6. Single Threaded CPU

```python
class Solution:
    def getOrder(self, tasks: List[List[int]]) -> List[int]:
        n = len(tasks)
        pairs = []
        for i in range(n):
            pairs.append((tasks[i][0], tasks[i][1], i))
        pairs.sort()

        time = 0
        min_heap = []
        i = 0
        ans = []

        while i < n or min_heap:
            # jump if idle
            if not min_heap and time < pairs[i][0]:
                time = pairs[i][0]

            # move the tasks to the min_heap for selection
            while i < n and time >= pairs[i][0]:
                start, processing, index = pairs[i]
                heappush(min_heap, (processing, index))
                i += 1

            # get an active task from heap
            processing, index = heappop(min_heap)
            time += processing
            ans.append(index)

        return ans
```

##### 8. Reorganize String

```python
class Solution:
    def reorganizeString(self, s: str) -> str:
        freq = Counter(s)
        max_heap = []
        for ch, cnt in freq.items():
            heappush(max_heap, (-cnt, ch))
        ans = ""

        while max_heap:
            cnt, ch = heappop(max_heap)
            if ans and ans[-1] == ch:
                if not max_heap:
                    return ""
                nxt_cnt,nxt_ch = heappop(max_heap)
                ans += nxt_ch
                nxt_cnt += 1

                if nxt_cnt < 0:
                    heappush(max_heap, (nxt_cnt,nxt_ch))
                heappush(max_heap, (cnt, ch))
            else:
                ans += ch
                cnt += 1
                if cnt < 0:
                    heappush(max_heap, (cnt,ch))
        return ans
```

##### 9. Car Pooling

```python
class Solution:
    def carPooling(self, trips: List[List[int]], capacity: int) -> bool:
        trips.sort(key = lambda x: (x[1], x[2]))
        min_heap = []
        curr = 0
        for passengers, start, end in trips:
            # free up some space
            while min_heap and start >= min_heap[0][0]:
                _, passengers_drop = heappop(min_heap)
                curr -= passengers_drop
            curr += passengers
            if curr > capacity:
                return False
            heappush(min_heap, (end, passengers))
        return True
```

##### 10. Median in a Data Stream (Important)

```python
class MedianFinder:

    def __init__(self):
        # two heaps - small heap (max heap) & large heap (min heap)
        self.small, self.large = [], []

    def addNum(self, num: int) -> None:
        heappush(self.small, -num)

        # every element in small is less than large
        if self.small and self.large and (-self.small[0]) > self.large[0]:
            val = heappop(self.small)
            heappush(self.large, -val)
        
        # size is uneven
        if len(self.small) > len(self.large) + 1:
            val = heappop(self.small)
            heappush(self.large, -val)
        if len(self.large) > len(self.small) + 1:
            val = heappop(self.large)
            heappush(self.small, -val)

    def findMedian(self) -> float:
        if len(self.small) > len(self.large):
            return -self.small[0]
        if len(self.large) > len(self.small):
            return self.large[0]
        return (-self.small[0] + self.large[0]) / 2
```

##### 1. Insert Interval

```python
class Solution:
    def insert(self, intervals: List[List[int]], new_interval: List[int]) -> List[List[int]]:
        n,i,ans = len(intervals),0,[]
        while i < n:
            if intervals[i][1] < new_interval[0]:
                # intervals[i] is before new_interval - no overlap
                ans.append(intervals[i])
            elif new_interval[1] < intervals[i][0]:
                # new_interval is before intervals[i] - no overlap
                ans.append(new_interval)
                new_interval = intervals[i]
            else:
                # overlap
                new_interval[0] = min(new_interval[0], intervals[i][0])
                new_interval[1] = max(new_interval[1], intervals[i][1])
            i += 1
        ans.append(new_interval)
        return ans
            
```

##### 2. Merge Intervals

```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        intervals.sort()
        ans = [intervals[0]]
        for i in range(1,len(intervals)):
            if intervals[i][0] > ans[-1][1]:
                ans.append(intervals[i])
            else:
                ans[-1][1] = max(ans[-1][1], intervals[i][1])
        return ans
```

##### 3. Non overlapping Intervals

```python
class Solution:
    def eraseOverlapIntervals(self, intervals: List[List[int]]) -> int:
        # Standard Problem
        # minimize the intervals to be removed
        # smaller intervals with early end time should be prioritized

        intervals.sort(key = (lambda x: x[1]))
        ans = 0
        temp = intervals[0][1]
        for i in range(1,len(intervals)):
            if intervals[i][0] < temp:
                # overlap
                ans += 1
            else:
                # no overlap
                temp = intervals[i][1]
        return ans
```

##### 4. Remove Interval

```python
class Solution:
    def removeInterval(self, intervals: List[List[int]], toBeRemoved: List[int]) -> List[List[int]]:
        remove_start, remove_end = toBeRemoved
        ans = []
        
        for start, end in intervals:
            if start > remove_end or end < remove_start:
                # no overlap
                ans.append([start,end])
            else:
                if start < remove_start:
                    ans.append([start,remove_start])
                if end > remove_end:
                    ans.append([remove_end,end])
        return ans
```

##### 6. Meeting Rooms II

```python
from heapq import heappush, heappop
class Solution:
    def minMeetingRooms(self, intervals: List[List[int]]) -> int:
        intervals.sort(key=(lambda x: (x[0], x[1])))

        min_heap = []
        for start, end in intervals:
            if min_heap and start >= min_heap[0]:
                # reuse same room
                heappop(min_heap)
            # Every meeting - a room
            heappush(min_heap, end)
        return len(min_heap)
```

##### 7. Meeting Rooms III

```python
class Solution:
    def mostBooked(self, n: int, meetings: List[List[int]]) -> int:
        meetings.sort(key = lambda x: (x[0],x[1]))
        unused_rooms = []
        for i in range(n):
            heappush(unused_rooms, i)
        
        used_count = [0] * n
        min_heap = []
    
        for start,end in meetings:
            # free some meeting rooms
            while min_heap and start >= min_heap[0][0]:
                _, room = heappop(min_heap)
                heappush(unused_rooms, room)
            
            if unused_rooms:
                # rooms available
                room = heappop(unused_rooms)
                used_count[room] += 1
                heappush(min_heap, (end, room))
            else:
                # rooms not available
                available_at, room = heappop(min_heap)
                used_count[room] += 1
                heappush(min_heap, (available_at + end - start, room))
        
        maxi, ans = -1, -1
        for i in range(n):
            if used_count[i] > maxi:
                maxi = used_count[i]
                ans = i
        return ans
```

##### 8. Minimum Interval to Include Each Query

```python
from heapq import heappush, heappop
class Solution:
    def minInterval(self, intervals: List[List[int]], queries: List[int]) -> List[int]:
        intervals.sort()

        sorted_queries = sorted((q, i) for i, q in enumerate(queries))
        ans = [-1] * len(queries)

        min_heap = []
        i, n = 0, len(intervals)

        for q, idx in sorted_queries:

            while i < n and intervals[i][0] <= q:
                start, end = intervals[i]
                heappush(min_heap,(end-start+1,end))
                i += 1
            
            while min_heap and min_heap[0][1] < q:
                heappop(min_heap)
            
            if min_heap:
                ans[idx] = min_heap[0][0]
        return ans
```
