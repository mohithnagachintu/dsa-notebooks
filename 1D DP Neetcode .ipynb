##### 3. House Robber


```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        # dp = [-1 for _ in range(n)]
        # def solve(i):
        #     if i >= n:
        #         return 0
        #     if dp[i] != -1: 
        #         return dp[i]
        #     # two choices
        #     a = solve(i+2) + nums[i]
        #     b = solve(i+1)
        #     dp[i] = max(a,b)
        #     return dp[i]
        # return solve(0)

        dp = [0] * (n+2)
        for i in range(n-1,-1,-1):
            a = dp[i+2] + nums[i]
            b = dp[i+1] 
            dp[i] = max(a,b)
        return dp[0]
```

##### 4. House Robber II

```python
class Solution:
    def _rob(self, nums: List[int]) -> int:
        n = len(nums)
        # dp = [-1 for _ in range(n)]
        # def solve(i):
        #     if i >= n:
        #         return 0
        #     if dp[i] != -1: 
        #         return dp[i]
        #     # two choices
        #     a = solve(i+2) + nums[i]
        #     b = solve(i+1)
        #     dp[i] = max(a,b)
        #     return dp[i]
        # return solve(0)

        dp = [0] * (n+2)
        for i in range(n-1,-1,-1):
            a = dp[i+2] + nums[i]
            b = dp[i+1] 
            dp[i] = max(a,b)
        return dp[0]

    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        if n == 1:
            return nums[0]
        elif n == 2:
            return max(nums[0],nums[1])
        return max(self._rob(nums[:-1]), self._rob(nums[1:]))
```

##### 6. Paint Fence

```python
class Solution:
    def numWays(self, n: int, k: int) -> int:
        dp = [-1] * (n+1)

        def solve(i):
            if i==1:
                return k
            elif i==2:
                return k * k
            if dp[i] != -1:
                return dp[i]

            # Use a different color than i-1, I have (k-1) choices to paint i
            a = solve(i-1) * (k-1)
            
            # use the same color as (i-1), but (i-1) should be painted with different color than (i-2)
            # To paint (i-2) we have (k-1) choices
            b = 1 * (k-1) * solve(i-2)
            dp[i] = a + b
            return dp[i]

        return solve(n)
```

##### 8. Handshakes That don't cross

```python
class Solution:
    def numberOfWays(self, n: int) -> int:
        # let n = 2i
        # Assume i is fixed and shook hand with some k,  it splits to left and right 
        # left = 2j, right = 2j-2i-2,
        # left shake hands in dp[j] ways, and right shake hands in dp[i-j-1]
        dp = [0] * (n//2+1)
        dp[0] = 1
        m = 1000000007
        for i in range(1,n//2+1):
            for j in range(i):
                dp[i] += (dp[j] * dp[i-j-1])
                dp[i] %= m
        return dp[-1]
```

##### 9. Palindromic Substrings

```python
class Solution:
    def countSubstrings(self, s: str) -> int:
        n = len(s)

        dp = [[-1 for _ in range(n)] for _ in range(n)]

        def solve(i,j):
            if i >= j:
                return 1
            if dp[i][j] != -1:
                return dp[i][j]
            
            ans = 0
            if s[i] == s[j]:
                ans = solve(i+1,j-1)
            dp[i][j] = ans
            return ans

        ans = 0
        for i in range(n):
            for j in range(i,n):
                ans += solve(i,j)
        return ans
```
