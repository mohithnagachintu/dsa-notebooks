# Working with String Data

```python
s1 = "I am an Immutable String"
print(s1[3])
```

```python
print(s1[8:17])
```

```python
# s1[8:17]="Mutable"         Error. Can't be changed inplace
```

```python
print(s1,id(s1))
s1 = s1[:6]+s1[7:8]+s1[10:17]+s1[17:]
print(s1, id(s1))
```
