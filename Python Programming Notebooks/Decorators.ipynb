```python
def dec(orig):
    def wrapper(*args,**kwargs):
        print("Before executing the original function..")
        orig(*args,**kwargs)
        print("After executing the oringal function...")
    return wrapper
```

```python
def orig(name,age):
    print(name,age)
```

```python
orig('Hello',23)
```

```python
@dec
def orig1(name,age):
    print(name,age)
```

```python
orig1('Hello',23)
```

```python
def time_cal(func):
    import time
    def wrapper(*args, **kwargs):
        t1 = time.time()
        func(*args, **kwargs)
        t2 = time.time() 
        print('Execution time; ', t2-t1)
    return wrapper
```

```python
@time_cal
def add_numbers(n):
    s = 0
    for i in range(1,n+1):
        s += i
    return s
```

```python
add_numbers(100000)
```

```python
add_numbers(5000000)
```

### Calculate time

```python
def time_cal(orig):
    import time
    def wrapper(*args,**kwargs):
        t1 = time.time()
        orig(*args,**kwargs)
        t2 = time.time() - t1
        print('Execution time: ',t2)
    return wrapper
```

```python
@time_cal
def sumN(n):
    s = 0
    for i in range(1,n):
        s += i
    print(f'Sum of {n} numbers is {s}')
```

```python
sumN(2000)
```
