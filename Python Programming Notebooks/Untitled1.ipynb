```python
n = int(input())
while n != 0:
    lst = input().split()
    lst = [int(x) for x in lst]
    print(max(lst))
    n = n-1 
```

```python
import math
n = int(input())
while n!=0:
    x = int(input())
    print(math.sqrt(x))
```

```python
n = int(input())
while n != 0:
    lst = input().split()
    lst = [int(x) for x in lst]
    lst.sort()
    print(lst[-2])
    n = n-1
```

```python
n = int(input())
i = 2
while n>0:
    f = [x for x in range(2,i+1) if i%x == 0 ]
    if len(f)==1:
        print(i)
        n = n-1
    i = i+1
```
