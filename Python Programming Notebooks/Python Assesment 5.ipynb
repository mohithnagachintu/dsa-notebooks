```python
class UnderFlowError(Exception):
    def __init__(self,msg):
        self.err_msg = msg
    def __str__(self):
        return "Underflow Error : " + self.err_msg
```

```python
class Share:
    def __init__(self,name,shares,uprice):
        self.__CName = name
        self.__NShares = shares
        self.__Uprice = uprice
    @property
    def CName(self):
        return self.__CName
    @CName.setter
    def CName(self,value):
        self.__CName = value
    @property
    def NShares(self):
        return self.__NShares
    @NShares.setter
    def NShares(self,value):
        self.__NShares = value
    @property
    def Uprice(self):
        return self.__Uprice
    @Uprice.setter
    def Uprice(self,value):
        self.__Uprice = value

    def __getitem__(self,index):
        if isinstance(index,int):
            if index == 0:
                return self.CName
            elif index == 1:
                return self.Nshares
            elif index == 2:
                return self.Uprice
            else :
                raise IndexError
    def __setitem__(self,index,value):
        if isinstance(index,int):
            if index == 0:
                self.CName = value
            elif index == 1:
                self.NShares = value
            elif index == 2:
                self.Uprice = value
            else:
                raise IndexError
    def __add__(self,other):
        T = Share(self.CName,0,self.Uprice)
        if isinstance(other,int):
            T.NShares = self.NShares + other
            return T
        else :
            return None
    def __radd__(self,other):
        T = Share(self.CName,0,self.Uprice)
        if isinstance(other,int):
            T.NShares = self.NShares + other
            return T
        else :
            return None
    def __iadd__(self,other):
        if isinstance(other,int):
            self.NShares = self.NShares + other
            return self
        else :
            return None
    def __sub__(self,other):
        T = Share(self.CName,0,self.Uprice)
        if isinstance(other,int):
            if other <= self.NShares:
                T.NShares = self.NShares - other
                return T
            else:
                raise UnderFlowError("No. of shares are less than the value to be deducted" )
        else:
            return None
    def __isub__(self,other):
        if isinstance(other,int):
            self.NShares = self.NShares - other
            return self
        else:
            return None
    def __str__(self):
        return "(" + self.CName + "," + str(self.NShares) + ","+str(self.Uprice)+")\n"
    def __repr__(self):
        return "(" + self.CName + "," + str(self.NShares) + ","+str(self.Uprice)+")\n"
```

```python
class Person:
    def __init__(self,name,amount):
        self.PName = name
        self.share_amount = amount
    def purchaseShares(self,share1,sh_pur):
        if share1.NShares < sh_pur:
            raise UnderFlowError("No. of shares available by company are less")
        else:
            self.share_amount = self.share_amount + share1.Uprice * sh_pur
            share1.NShares -= sh_pur
    
    def sellShares(self,share2,sh_sol):
        t = self.share_amount - share2.Uprice * sh_sol 
        if t<0:
            raise UnderFlowError("Share Amount of the person is less ")
        else :
            self.share_amount = self.share_amount - share2.Uprice * sh_sol
            share2.NShares += sh_sol
    
    def __str__(self):
        return "("+self.PName + ","+str(self.share_amount)+")"
    def __repr__(self):
        return "("+self.PName + ","+str(self.share_amount)+")"
        
```
