### Expand from Center — Template

```python
def expand(s, left, right):
    while left >= 0 and right < len(s) and s[left] == s[right]:
        left -= 1
        right += 1
    return left + 1, right - 1

def longest_palindrome(s):
    start = end = 0
    for i in range(len(s)):
        for l, r in (expand(s, i, i), expand(s, i, i + 1)):
            if r - l > end - start:
                start, end = l, r
    return s[start:end + 1]
```

````python
def countSubstrings(self, s: str) -> int:
    ans, n = 0, len(s)

    def count_palindrome_expand_center(left, right):
        ans = 0
        while left >= 0 and right < n:
            if s[left] != s[right]:
                break
            left -= 1
            right += 1
            ans += 1
        return ans
    
    for i in range(n):
        # odd length palindromes
        ans += count_palindrome_expand_center(i, i)
        # even length palindromes
        ans += count_palindrome_expand_center(i, i+1)
    
    return ans
````
