```python
def singleNumberIII(nums):
    xor_all = 0
    for num in nums: 
        xor_all ^= num
    # get the rightmost set bit
    # Formula X = (xor_all & (xor_all-1)) ^ xor_all
    x = (xor_all & (xor_all-1)) ^ xor_all
    print(x)
    # seperate numbers to two buckets, Don't need to store them
    b1, b2 = 0,0
    for num in nums:
        if x & num: 
            b1 ^= num
        else:
            b2 ^= num
    return [b1,b2]
singleNumberIII([1,2,1,3,2,5])
```

###### A list(array) of words will be provided, I need to find a pair of words where there is no common letter in between and return the maximum product of their length possible.
Examples : [”dog”, “cat”, “pull”, “space”, ”feed”] ⇒ answer is “pull” and “feed”

```python
def solve(words):
    n = len(words)
    word_bitmasks = [None] * n
    for i in range(n):
        bitmask = 0
        for ch in words[i]:
            bitmask |= 1 << (ord(ch)-ord('a'))
        word_bitmasks[i] = bitmask
    ans = 0
    for i in range(n):
        for j in range(i+1,n):
            if word_bitmasks[i] & word_bitmasks[j] == 0:
                if len(words[i]) * len(words[j]) > ans:
                    ans = len(words[i]) * len(words[j])
                    print(words[i], words[j])
                # ans = max(ans, len(words[i]) * len(words[j]))
    return ans
```

```python
solve(['dog', 'cat', 'pull', 'space', 'feed'])
```

```python
# Another solution using Tries
class TrieNode:
    def __init__(self):
        self.children = {}
        self.word_lengths = []

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word, length):
        node = self.root
        for char in word:
            if char not in node.children:
                node.children[char] = TrieNode()
            node = node.children[char]
        node.word_lengths.append(length)

    def search(self):
        return self._search_helper(self.root, [])

    def _search_helper(self, node, word_lengths):
        max_product = 0

        # Compare all word lengths stored at this node with previous word lengths
        for length in node.word_lengths:
            for other_length in word_lengths:
                max_product = max(max_product, length * other_length)

        # Update word_lengths for child nodes
        current_word_lengths = word_lengths + node.word_lengths

        # Recursively search through the children nodes
        for child in node.children.values():
            max_product = max(max_product, self._search_helper(child, current_word_lengths))

        return max_product

def maxProduct(words):
    # Create the Trie
    trie = Trie()

    # Insert all words into the trie
    for word in words:
        sorted_word = ''.join(sorted(set(word)))  # Sort letters in the word and remove duplicates
        trie.insert(sorted_word, len(word))

    # Search for the maximum product
    return trie.search()

# Example usage:
words = ["dog", "cat", "pull", "space", "feed"]
result = maxProduct(words)
print("Maximum product:", result)
```
