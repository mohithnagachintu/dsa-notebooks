# String Trie

#### Trie Implementation - Type 1
* insert
* search
* startsWith

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:

    def __init__(self):
        self.root = TrieNode()

    def insert(self, word: str) -> None:
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
        cur.end = True

    def search(self, word: str) -> bool:
        cur = self.root
        for c in word:
            if c not in cur.children:
                return False
            cur = cur.children[c]
        return cur.end
        
    def startsWith(self, prefix: str) -> bool:
        cur = self.root
        for c in prefix:
            if c not in cur.children:
                return False
            cur = cur.children[c]
        return True

# Your Trie object will be instantiated and called as such:
# obj = Trie()
# obj.insert(word)
# param_2 = obj.search(word)
# param_3 = obj.startsWith(prefix)
```

##### Implementation - Trie Part II 

* insert
* countWordsEqualTo
* countStartingWith
* erase

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = True
        self.ends_with = 0
        self.ch_count = 0

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
            cur.ch_count += 1

        cur.end = True
        cur.ends_with += 1

    def countWordsEqualTo(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                return 0
            cur = cur.children[c]
        return cur.ends_with
    
    def countWordsStartingWith(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                return 0
            cur = cur.children[c]
        return cur.ch_count

    def erase(self, word):
        cur = self.root
        for c in word:
            if c in cur.children:
                cur = cur.children[c]
                cur.ch_count -= 1
            else:
                return
        cur.ends_with -= 1
        
```

##### 4. Number of Distinct Sub strings

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        cur = self.root
        new_nodes = 0

        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
                new_nodes += 1
            cur = cur.children[c]
        cur.end = True
        return new_nodes

class Solution:
    def countDistinctSubstring(self, s):
        n = len(s)
        trie = Trie()
        ans = 0
        for i in range(n):
            ans += trie.insert(s[i:])
        return ans + 1                
```

##### 5. Count Prefix and Suffix Pairs - II (Hard)

```python
class PairTrieNode:
    def __init__(self):
        # key =  (char_start, char_end)
        self.children = dict()
        self.count = 0

class PairTrie:
    def __init__(self):
        self.root = PairTrieNode()

    def insert_and_count(self,word):
        cur = self.root
        matches = 0
        l = len(word)
        for i in range(l):
            pair = (word[i],word[l-i-1])
            if pair not in cur.children:
                cur.children[pair] = PairTrieNode()
            cur = cur.children[pair]
            matches += cur.count
        cur.count += 1
        return matches

class Solution:
    def countPrefixSuffixPairs(self, words: List[str]) -> int:
        pair_trie = PairTrie()
        ans = 0
        for word in words:
            ans += pair_trie.insert_and_count(word)
        return ans
```

##### 7. Extra Characters in a String (2707)

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()
    
    def insert(self,word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
        cur.end = True

class Solution:
    def minExtraChar(self, s: str, dictionary: List[str]) -> int:
        # Similar as Word break - Trie Optimization one.
        n = len(s)
        trie = Trie()
        for word in dictionary:
            trie.insert(word)
        
        dp = [0] * (n+1)
        dp[n] = 0

        for i in range(n-1,-1,-1):
            # assume s[i] as extra character
            dp[i] = 1 + dp[i+1]

            cur = trie.root
            for j in range(i,n):
                if s[j] not in cur.children:
                    break
                cur = cur.children[s[j]]
                if cur.end:
                    dp[i] = min(dp[i], dp[j+1])

        return dp[0]
```

##### 8. Sum of Prefix Scores of Strings (2416)

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.count = 0

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self,word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
            cur.count += 1

    def get_count(self,word):
        cur = self.root
        ans = 0
        for c in word:
            cur = cur.children[c]
            ans += cur.count
        return ans

class Solution:
    def sumPrefixScores(self, words: List[str]) -> List[int]:
        trie = Trie()
        for word in words:
            trie.insert(word)
        ans = []
        for word in words:
            ans.append(trie.get_count(word))
        return ans
```

##### 9. Word Search II - Important (Amazon Question)

###### a. Trie Implementation

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False
    
class Trie:
    def __init__(self):
        self.root = TrieNode()
    
    def insert(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
        cur.end = True

class Solution:
    def findWords(self, board: List[List[str]], words: List[str]) -> List[str]:
        trie = Trie()
        for word in words: # O(W * L)
            trie.insert(word)

        m,n = len(board), len(board[0])
        ans = set()
        def dfs(i,j,node,path):
            c = board[i][j]
            if c not in node.children:
                return
            node = node.children[c]
            path += c
            if node.end:
                ans.add(path)
            board[i][j] = '#'
            for (di,dj) in [(0,-1),(0,1), (1,0), (-1,0)]:
                new_i, new_j = i + di, j + dj
                if new_i >= 0 and new_i < m\
                    and new_j >=0 and new_j < n\
                    and board[new_i][new_j] != '#':
                    dfs(new_i,new_j,node,path)
            board[i][j] = c
        for i in range(m):
            for j in range(n):
                dfs(i,j,trie.root,'')  # (4^L)
        return list(ans)
        # TC = O(W*L) + O(M*N*4^L)
        
        
```

###### b. Pruned Trie Optimization

```python
# Trie (Pruned) Optimization
# 1. Ignore multiple utterances of a word

class TrieNode:
    def __init__(self):
        self.children = dict()
        self.word = None
    
class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self,word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
        cur.word = word

class Solution:
    def findWords(self, board: List[List[str]], words: List[str]) -> List[str]:
        trie = Trie()
        for word in words:
            trie.insert(word)
        
        m,n = len(board), len(board[0])
        ans = []

        def dfs(i,j,parent,c):
            node = parent.children[c]

            # optimization - Find and immediately 'unset' to avoid duplicates
            if node.word:
                ans.append(node.word)
                node.word = None
            
            board[i][j] = '#'
            for (di,dj) in [(0,-1),(0,1), (1,0), (-1,0)]:
                new_i, new_j = i + di, j + dj
                if 0 <= new_i < m and 0 <= new_j < n\
                    and board[new_i][new_j] != '#':
                    if board[new_i][new_j] in node.children:
                        dfs(new_i,new_j,node, board[new_i][new_j])
            board[i][j] = c

        for i in range(m):
            for j in range(n):
                if board[i][j] in trie.root.children:
                    dfs(i,j,trie.root,board[i][j])
        return ans
```

##### 11. Remove Sub-Folders from the Filesystem

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False
    
class Trie:
    def __init__(self):
        self.root = TrieNode()
        
    
    def insert(self, absolute_path):
        sub_paths = absolute_path.split('/')
        cur = self.root
        for path in sub_paths:
            if path not in cur.children:
                cur.children[path] = TrieNode()
            cur = cur.children[path]
        cur.end = True

    def remove_helper(self, absolute_path):
        sub_paths = absolute_path.split('/')
        cur = self.root
        for path in sub_paths:
            if path in cur.children and cur.children[path].end == True:
                cur.children[path] = TrieNode()
                cur.children[path].end = True
    
    def get_absolute_paths(self):
        absolute_paths = []
        def dfs(cur, word):
            if cur.end == True:
                absolute_paths.append(word[1:])
                return
            for key in cur.children:
                dfs(cur.children[key],f'{word}/{key}')
        dfs(self.root, '')
        return absolute_paths

class Solution:
    def removeSubfolders(self, folder: List[str]) -> List[str]:
        # trie = Trie()
        # for word in folder:
        #     trie.insert(word)
        # for word in folder:
        #     trie.remove_helper(word)
        # return trie.get_absolute_paths()

        # Optimization
        folder.sort()
        root = TrieNode()
        ans = []
        for path in folder:
            parts = path.split('/')
            cur = root
            flag = False
            for i in range(1,len(parts)):
                part = parts[i]
                if part not in cur.children:
                    cur.children[part] = TrieNode()
                cur = cur.children[part]
                if cur.end:
                    flag = True
                    break
            if not flag:
                cur.end = True
                ans.append(path)
        return ans

```

##### 12. Design File System

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.value = -1

class FileSystem:

    def __init__(self):
        self.root = TrieNode()

    def createPath(self, path: str, value: int) -> bool:
        parts = path.split('/')[1:]
        n = len(parts)
        cur = self.root
        for i in range(n-1):
            part = parts[i]
            if part not in cur.children:
                return False
            cur = cur.children[part]
        if parts[n-1] in cur.children:
            return False
        cur.children[parts[n-1]] = TrieNode()
        cur = cur.children[parts[n-1]]
        cur.value = value
        return True

    def get(self, path: str) -> int:
        cur = self.root
        parts = path.split('/')[1:]
        for part in parts:
            if part not in cur.children:
                return -1
            cur = cur.children[part]
        return cur.value

# Your FileSystem object will be instantiated and called as such:
# obj = FileSystem()
# param_1 = obj.createPath(path,value)
# param_2 = obj.get(path)
```

##### 13. Design In-Memory File System

```python
class TrieNode():
    def __init__(self,is_folder = True, file_content = None):
        self.children = dict()
        self.is_folder = is_folder
        self.file_content = file_content

class FileSystem:

    def __init__(self):
        self.root = TrieNode()

    def ls(self, path: str) -> List[str]:
        cur = self.root
        if path == "/":
            return sorted(cur.children.keys())
        parts = path.split('/')[1:]
        for part in parts:
            cur = cur.children[part]
        if not cur.is_folder:
            return [parts[-1]]
        return sorted(cur.children.keys())

    def mkdir(self, path: str) -> None:
        cur = self.root
        parts = path.split('/')[1:]
        for part in parts:
            if part not in cur.children:
                cur.children[part] = TrieNode()
            cur = cur.children[part]
        

    def addContentToFile(self, filePath: str, content: str) -> None:
        cur = self.root
        parts = filePath.split('/')[1:]
        n = len(parts)
        for i in range(n-1):
            cur = cur.children[parts[i]]
        file_name = parts[-1]
        if file_name not in cur.children:
            cur.children[file_name] = TrieNode(False,"")
        cur.children[file_name].file_content += content
    
    def readContentFromFile(self, filePath: str) -> str:
        cur = self.root
        parts = filePath.split('/')[1:]
        n = len(parts)
        file_name = parts[n-1]
        for i in range(n-1):
            cur = cur.children[parts[i]]
        return cur.children[file_name].file_content
        


# Your FileSystem object will be instantiated and called as such:
# obj = FileSystem()
# param_1 = obj.ls(path)
# obj.mkdir(path)
# obj.addContentToFile(filePath,content)
# param_4 = obj.readContentFromFile(filePath)
```
