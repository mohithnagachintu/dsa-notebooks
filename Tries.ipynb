#### Trie Implementation - Type 1
* insert
* search
* startsWith

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:

    def __init__(self):
        self.root = TrieNode()

    def insert(self, word: str) -> None:
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
        cur.end = True

    def search(self, word: str) -> bool:
        cur = self.root
        for c in word:
            if c not in cur.children:
                return False
            cur = cur.children[c]
        return cur.end
        
    def startsWith(self, prefix: str) -> bool:
        cur = self.root
        for c in prefix:
            if c not in cur.children:
                return False
            cur = cur.children[c]
        return True

# Your Trie object will be instantiated and called as such:
# obj = Trie()
# obj.insert(word)
# param_2 = obj.search(word)
# param_3 = obj.startsWith(prefix)
```

##### Implementation - Trie Part II 

* insert
* countWordsEqualTo
* countStartingWith
* erase

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = True
        self.ends_with = 0
        self.ch_count = 0

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
            cur.ch_count += 1

        cur.end = True
        cur.ends_with += 1

    def countWordsEqualTo(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                return 0
            cur = cur.children[c]
        return cur.ends_with
    
    def countWordsStartingWith(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                return 0
            cur = cur.children[c]
        return cur.ch_count

    def erase(self, word):
        cur = self.root
        for c in word:
            if c in cur.children:
                cur = cur.children[c]
                cur.ch_count -= 1
            else:
                return
        cur.ends_with -= 1
        
```

##### 3. Longest Word with all prefixes - https://www.naukri.com/code360/problems/complete-string_2687860

```python
from sys import *
from collections import *
from math import *

from typing import *

class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
        cur.end = True
    

def is_complete_string(trie, word):
    cur = trie.root
    for c in word:
        if c not in cur.children:
            return False
        cur = cur.children[c]
        if cur.end == False:
            return False
    return True

def completeString(n: int, words: List[str])-> str:
    trie = Trie()
    for word in words:
        trie.insert(word)
    ans = ''
    for word in words:
        if is_complete_string(trie,word):
            if len(word) > len(ans) or (len(word) == len(ans) and word < ans):
                ans = word
    if len(ans): 
        return ans
    return 'None'
```

##### 4. Number of Distinct Sub strings

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        cur = self.root
        new_nodes = 0

        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
                new_nodes += 1
            cur = cur.children[c]
        cur.end = True
        return new_nodes

class Solution:
    def countDistinctSubstring(self, s):
        n = len(s)
        trie = Trie()
        ans = 0
        for i in range(n):
            ans += trie.insert(s[i:])
        return ans + 1                
```

##### 5. Count Prefix and Suffix Pairs - II (Hard)

```python
class PairTrieNode:
    def __init__(self):
        # key =  (char_start, char_end)
        self.children = dict()
        self.count = 0

class PairTrie:
    def __init__(self):
        self.root = PairTrieNode()

    def insert_and_count(self,word):
        cur = self.root
        matches = 0
        l = len(word)
        for i in range(l):
            pair = (word[i],word[l-i-1])
            if pair not in cur.children:
                cur.children[pair] = PairTrieNode()
            cur = cur.children[pair]
            matches += cur.count
        cur.count += 1
        return matches

class Solution:
    def countPrefixSuffixPairs(self, words: List[str]) -> int:
        pair_trie = PairTrie()
        ans = 0
        for word in words:
            ans += pair_trie.insert_and_count(word)
        return ans
```

##### 7. Extra Characters in a String (2707)

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()
    
    def insert(self,word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
        cur.end = True

class Solution:
    def minExtraChar(self, s: str, dictionary: List[str]) -> int:
        # Similar as Word break - Trie Optimization one.
        n = len(s)
        trie = Trie()
        for word in dictionary:
            trie.insert(word)
        
        dp = [0] * (n+1)
        dp[n] = 0

        for i in range(n-1,-1,-1):
            # assume s[i] as extra character
            dp[i] = 1 + dp[i+1]

            cur = trie.root
            for j in range(i,n):
                if s[j] not in cur.children:
                    break
                cur = cur.children[s[j]]
                if cur.end:
                    dp[i] = min(dp[i], dp[j+1])

        return dp[0]
```
