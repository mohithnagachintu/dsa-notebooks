#### Trie Implementation - Type 1
* insert
* search
* startsWith

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:

    def __init__(self):
        self.root = TrieNode()

    def insert(self, word: str) -> None:
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
        cur.end = True

    def search(self, word: str) -> bool:
        cur = self.root
        for c in word:
            if c not in cur.children:
                return False
            cur = cur.children[c]
        return cur.end
        
    def startsWith(self, prefix: str) -> bool:
        cur = self.root
        for c in prefix:
            if c not in cur.children:
                return False
            cur = cur.children[c]
        return True

# Your Trie object will be instantiated and called as such:
# obj = Trie()
# obj.insert(word)
# param_2 = obj.search(word)
# param_3 = obj.startsWith(prefix)
```

##### Implementation - Trie Part II 

* insert
* countWordsEqualTo
* countStartingWith
* erase

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = True
        self.ends_with = 0
        self.ch_count = 0

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
            cur.ch_count += 1

        cur.end = True
        cur.ends_with += 1

    def countWordsEqualTo(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                return 0
            cur = cur.children[c]
        return cur.ends_with
    
    def countWordsStartingWith(self, word):
        cur = self.root
        for c in word:
            if c not in cur.children:
                return 0
            cur = cur.children[c]
        return cur.ch_count

    def erase(self, word):
        cur = self.root
        for c in word:
            if c in cur.children:
                cur = cur.children[c]
                cur.ch_count -= 1
            else:
                return
        cur.ends_with -= 1
        
```

##### 4. Number of Distinct Sub strings

```python
class TrieNode:
    def __init__(self):
        self.children = dict()
        self.end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        cur = self.root
        new_nodes = 0

        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
                new_nodes += 1
            cur = cur.children[c]
        cur.end = True
        return new_nodes

class Solution:
    def countDistinctSubstring(self, s):
        n = len(s)
        trie = Trie()
        ans = 0
        for i in range(n):
            ans += trie.insert(s[i:])
        return ans + 1                
```
