##### 1. BFS and 2. DFS

```python
class Solution:
    def dfsOfGraph(self, V, adj):
        dfs_order = []
        vis = [False] * V

        def dfs(node):
          vis[node] = True
          dfs_order.append(node)
          for adj_node in adj[node]:
            if not vis[adj_node]:
              dfs(adj_node)
          
        for v in range(V):
          if not vis[v]:
            dfs(v)
        return dfs_order

    
    def bfsOfGraph(self, V, adj):
      bfs_order = []
      vis = [False] * V

      def bfs(start_node):
        q = deque()
        q.append(start_node)
        vis[start_node] = True

        while len(q):
          node = q.popleft()
          bfs_order.append(node)
          for adj_node in adj[node]:
            if not vis[adj_node]:
              q.append(adj_node)
              vis[adj_node] = True
      for v in range(V):
        if not vis[v]:
          bfs(v)
      return bfs_order
```

```python
class DisjointSet:
    def __init__(self):
        self.parent = dict()
        self.rank = dict()
        self.size = dict()
    
    def find(self,node):
        if node not in self.parent:
            self.parent[node] = node
            self.rank[node] = 0
            self.size[node] = 1

        if node != self.parent[node]:
            self.parent[node] = self.find(self.parent[node])
        return self.parent[node]
    
    def union(self, u, v, by_size = True):
        root_u = self.find(u)
        root_v = self.find(v)
        if root_u == root_v:
            return

        if by_size:
            if self.size[root_u] < self.size[root_v]:
                self.parent[root_u] = root_v
                self.size[root_v] += self.size[root_u]
            else:
                self.parent[root_v] = root_u
                self.size[root_u] += self.size[root_v]
        else:
            if self.rank[root_u] < self.rank[root_v]:
                self.parent[root_u] = root_v
            elif self.rank[root_u] > self.rank[root_v]:
                self.parent[root_v] = root_u
            else:
                self.parent[root_v] = root_u
                self.rank[root_u] += 1

    def connected(self, u, v):
        return self.find(u) == self.find(v)


class Solution:
    def findCircleNum(self, isConnected: List[List[int]]) -> int:
        # n, ans = len(isConnected),0
        # vis = [False] * n
        # def dfs(i):
        #     vis[i] = True
        #     for j in range(n):
        #         if isConnected[i][j] and (not vis[j]):
        #             dfs(j)
        # def bfs(i):
        #     q = deque()
        #     q.append(i)
        #     vis[i] = True
        #     while q:
        #         node = q.popleft()
        #         for j in range(n):
        #             if isConnected[node][j] and (not vis[j]):
        #                 q.append(j)
        #                 vis[j] = True
        # for i in range(n):
        #     if not vis[i]:
        #         ans += 1
        #         # dfs(i)
        #         bfs(i)
        # return ans

        ds = DisjointSet()
        n = len(isConnected)

        for i in range(n):
            for j in range(i+1,n):
                if isConnected[i][j]:
                    ds.union(i,j)

        roots = set()
        for i in range(n):
            roots.add(ds.find(i))
        return len(roots)
```

##### 4. Rotten Oranges

```python

def orangesRotting( grid: List[List[int]]) -> int:
    # Multi source BFS
    m, n = len(grid), len(grid[0])
    q = deque()
    fresh = 0

    for i in range(m):
        for j in range(n):
            if grid[i][j] == 2:
                q.append((i, j, 0))  # store (x, y, time)
            elif grid[i][j] == 1:
                fresh += 1

    # Step 2: BFS
    directions = [[0,1],[1,0],[0,-1],[-1,0]]
    minutes = 0

    while q:
        x, y, time = q.popleft()
        for dx, dy in directions:
            nx, ny = x + dx, y + dy
            if 0 <= nx < m and 0 <= ny < n and grid[nx][ny] == 1:
                grid[nx][ny] = 2  # rot the fresh orange
                fresh -= 1
                q.append((nx, ny, time + 1))
                minutes = time + 1

    return minutes if fresh == 0 else -1
```

##### 5. Flood Fill

```python
class Solution:
    def floodFill(self, grid: List[List[int]], sr: int, sc: int, color: int) -> List[List[int]]:
        m, n = len(grid), len(grid[0])

        q = deque()
        og_color = grid[sr][sc]
        if og_color == color:
            return grid

        grid[sr][sc] = color
        q.append((sr,sc))

        dij = [[-1,0],[1,0],[0,-1],[0,1]]

        while q:
            x,y = q.popleft()
            for dx,dy in dij:
                new_x, new_y = x+dx, y+dy
                if 0 <= new_x < m and 0 <= new_y < n and grid[new_x][new_y] == og_color:
                    q.append((new_x,new_y))
                    grid[new_x][new_y] = color
                
        return grid
```

##### 6. Cycle Detection in Undirected graph

```python
from collections import deque
class Solution:
    def isCycle(self, V, edges):
        adj_graph = [[] for _ in range(V)]
        for u,v in edges:
            adj_graph[u].append(v)
            adj_graph[v].append(u)
    
        vis = [False] * V
		
        def detect_cycle_bfs(start_node):
            q = deque()
            q.append((start_node,-1))
            vis[start_node] = True     
            while q:
                node, parent = q.popleft()
                for adj_node in adj_graph[node]:
                    if not vis[adj_node]:
                        q.append((adj_node, node))
                        vis[adj_node] = True
                    elif adj_node != parent:
                       # if any adj_node visited other than parent
                       # then cycle exists'
                       return True
            return False
        
        def detect_cycle_dfs(node,parent):
            vis[node] = True
            for adj_node in adj_graph[node]:
                if not vis[adj_node]:
                    if detect_cycle_dfs(adj_node,node):
                        return True
                elif adj_node != parent:
                    return True
            return False
	       
        for i in range(V):
            if not vis[i]:
                # if detect_cycle_bfs(i):
                if detect_cycle_dfs(i,-1):
                    return True
        return False
```

##### 7. Detecting a cycle in Directed graph

```python
class Solution:
    def isCyclic(self, V, edges):
        adj_graph = [[] for _ in range(V)]
        
        for u,v in edges:
            adj_graph[u].append(v)

        vis = [False] * V
        backtrack = [False] * V
        
        def dfs(node):
            vis[node] = True
            backtrack[node] = True
            
            for adj_node in adj_graph[node]:
                if not vis[adj_node]:
                    if dfs(adj_node):
                        return True
                elif backtrack[adj_node]:
                    return True
            backtrack[node] = False
            return False
        
        for i in range(V):
            if not vis[i]:
                if dfs(i):
                    return True
        return False
```

##### 8. 01 Matrix

```python
class Solution:
    def updateMatrix(self, mat: List[List[int]]) -> List[List[int]]:
        m,n = len(mat), len(mat[0])
        ans = [[0 for _ in range(n)] for _ in range(m)] 
        vis = [[False for _ in range(n)] for _ in range(m)]
        q = deque()
        for i in range(m):
            for j in range(n):
                if mat[i][j] == 0:
                    q.append((i,j,0))
                    vis[i][j] = True

        dij = [(0,1),(1,0),(0,-1),(-1,0)]
        while q:
            i,j,d = q.popleft()
            ans[i][j] = d
            for di,dj in dij:
                ni,nj = i+di,j+dj
                if 0<=ni<m and 0<=nj<n and not vis[ni][nj]:
                    q.append((ni,nj,d+1))
                    vis[ni][nj] = 1
        return ans
```

##### 10. No. of enclaves - Flood fill variation

```python
class Solution:
    def numEnclaves(self, grid: List[List[int]]) -> int:
        # same like flood fill
        m,n = len(grid), len(grid[0])
        def dfs(i,j):
            if i < 0 or i == m or j < 0 or j == n or grid[i][j] == 0:
                return
            grid[i][j] = 0
            dij = [(0,1),(1,0),(-1,0),(0,-1)]
            for di,dj in dij:
                ni,nj = i+di,j+dj
                dfs(ni,nj)

        for i in range(m):
            dfs(i,0)
            dfs(i,n-1)
        for j in range(n):
            dfs(0,j)
            dfs(m-1,j)
        
        ans = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j]:
                    ans += 1
        return ans
```

##### 11. No. of Distinct Islands

```python
class Solution:
    def numDistinctIslands(self, grid: List[List[int]]) -> int:
        m, n = len(grid), len(grid[0])
        vis = [[False for _ in range(n)] for _ in range(m)]
        unique_shapes = set()
        dij = [(0, 1), (1, 0), (0, -1), (-1, 0)]

        def dfs(i, j, base_i, base_j, shape):
            if (i < 0 or i >= m or j < 0 or j >= n or vis[i][j] or grid[i][j] == 0):
                return
            vis[i][j] = True
            shape.append((i - base_i, j - base_j))
            for di, dj in dij:
                ni, nj = i + di, j + dj
                dfs(ni, nj, base_i, base_j, shape)

        for i in range(m):
            for j in range(n):
                if grid[i][j] and not vis[i][j]:
                    island_shape = []
                    dfs(i, j, i, j, island_shape)
                    unique_shapes.add(tuple(island_shape))

        return len(unique_shapes)
```

##### 12. Number of Islands

```python
class DisjointSet:
    def __init__(self):
        self.parent = dict()
        self.rank = dict()
        self.size = dict()
    
    def find(self,node):
        if node not in self.parent:
            self.parent[node] = node
            self.rank[node] = 0
            self.size[node] = 1

        if node != self.parent[node]:
            self.parent[node] = self.find(self.parent[node])
        return self.parent[node]
    
    def union(self, u, v, by_size = True):
        root_u = self.find(u)
        root_v = self.find(v)
        if root_u == root_v:
            return

        if by_size:
            if self.size[root_u] < self.size[root_v]:
                self.parent[root_u] = root_v
                self.size[root_v] += self.size[root_u]
            else:
                self.parent[root_v] = root_u
                self.size[root_u] += self.size[root_v]
        else:
            if self.rank[root_u] < self.rank[root_v]:
                self.parent[root_u] = root_v
            elif self.rank[root_u] > self.rank[root_v]:
                self.parent[root_v] = root_u
            else:
                self.parent[root_v] = root_u
                self.rank[root_u] += 1

    def connected(self, u, v):
        return self.find(u) == self.find(v)

class Solution:
    def numIslands(self, grid: List[List[str]]) -> int:
        if not grid:
            return 0

        m,n = len(grid), len(grid[0])

        ds = DisjointSet()
        directions = [(1, 0), (0, 1)]  

        for r in range(m):
            for c in range(n):
                if grid[r][c] == '1':
                    for dr, dc in directions:
                        nr, nc = r + dr, c + dc

                        if (
                            0 <= nr < m and
                            0 <= nc < n and
                            grid[nr][nc] == '1'
                        ):
                            ds.union((r, c), (nr, nc))

        roots = set()

        for r in range(m):
            for c in range(n):
                if grid[r][c] == '1':
                    roots.add(ds.find((r, c)))

        return len(roots)
```

##### 13. Graph Diameter

```python
from collections import deque
class Solution:
    def diameter(self, V, edges):
        adj_graph = [[] for _ in range(V)]
        for u,v in edges:
            adj_graph[u].append(v)
            adj_graph[v].append(u)
          
        def bfs_depth_component(start_node):
            vis = [False] * V
            q = deque()
            q.append((start_node,0))
            vis[start_node] = True
            farthest_node, ans = start_node, 0
            while q:
                node, d = q.popleft()
                for adj_node in adj_graph[node]:
                    if not vis[adj_node]:
                        q.append((adj_node,d+1))
                        vis[adj_node] = True
                        if d + 1 > ans:
                            ans = d + 1
                            farthest_node = adj_node
            return ans, farthest_node
            
        # 1. Naive Approach - Run BFS from every node and find maximum
        # ans = 0
        # for i in range(V):
        #     dist, far_node = bfs_depth_component(i)
        #     ans = max(ans, dist)
        # return ans
        
        # 2. Optimal Approach - 2 BFS
        # Idea: 1st find a farthest node from any node.
        # 2nd from farthest node in step 1, get the farthest node and depth, that's the diameter
        depth1, farthest_node = bfs_depth_component(0)
        depth2, farthest_node = bfs_depth_component(farthest_node)
        return depth2
        
```

# Graph Coloring Problems

##### 1. Bipartite Graph

```python
class Solution:
    def isBipartite(self, graph: List[List[int]]) -> bool:
        n = len(graph)
        colors = [-1] * n

        def bfs_coloring(start_node):
            q = deque()
            q.append((start_node,0))
            colors[start_node] = 0

            while q:
                node, col = q.popleft()
                for adj_node in graph[node]:
                    if colors[adj_node] == -1:
                        q.append((adj_node, not col))
                        colors[adj_node] = not col
                    elif colors[adj_node] == col:
                        return False
            return True
        
        for i in range(n):
            if colors[i] == -1:
                if bfs_coloring(i) == False:
                    return False
        return True
```

##### 2. Possible Bipartition

```python
class Solution:
    def possibleBipartition(self, n: int, dislikes: List[List[int]]) -> bool:
        adj_graph = [[] for _ in range(n+1)]
        for u,v in dislikes:
            adj_graph[u].append(v)
            adj_graph[v].append(u)
    
        colors = [-1] * (n+1)
    
        def is_bipartite(start_node):
            q = deque()
            q.append((start_node,0))
            colors[start_node] = 0

            while q:
                node, col = q.popleft()
                for adj_node in adj_graph[node]:
                    if colors[adj_node] == -1:
                        q.append((adj_node, not col))
                        colors[adj_node] = not col
                    elif colors[adj_node] == col:
                        return False
            return True

        for i in range(n):
            if colors[i] == -1:
                if not is_bipartite(i):
                    return False
        return True
```

##### 3. Divide Nodes Into the Maximum Number of Groups (Hard)


```python
class Solution:
    def magnificentSets(self, n: int, edges: List[List[int]]) -> int:
        adj_graph = [[] for _ in range(n+1)]
        for u,v in edges:
            adj_graph[u].append(v)
            adj_graph[v].append(u)
        
        # 1. check bipartite
        # 2. for every component,
        #    find maximum bfs depth possible
        colors = [-1] * (n+1)
        def is_bipartite(start_node):
            q = deque()
            q.append((start_node,0))
            colors[start_node] = 0
            component_nodes = []
            while q:
                node, col = q.popleft()
                component_nodes.append(node)
                for adj_node in adj_graph[node]:
                    if colors[adj_node] == -1:
                        q.append((adj_node, not col))
                        colors[adj_node] = not col
                    elif colors[adj_node] == col:
                        return False,[]
            return True,component_nodes

        def bfs_depth_component(start_node):
            vis = [False] * (n+1)
            q = deque()
            q.append((start_node,0))
            vis[start_node] = True
            farthest_node, ans = start_node, 0
            while q:
                node, d = q.popleft()
                for adj_node in adj_graph[node]:
                    if not vis[adj_node]:
                        q.append((adj_node,d+1))
                        vis[adj_node] = True
                        if d + 1 > ans:
                            ans = d + 1
                            farthest_node = adj_node
            return ans, farthest_node

        total_components = 0
        for i in range(1,n+1):
            if colors[i] == -1:
                possible, components = is_bipartite(i)
                if not possible:
                    return -1
                best_depth = 0
                for node in components:
                    depth, _ = bfs_depth_component(node)
                    best_depth = max(best_depth, depth)
                total_components += best_depth + 1  
        return total_components
```

##### 4. M-Coloring Problem

```python
class Solution:
    def graphColoring(self, V, edges, m):
        adj_graph = [[] for _ in range(V)]
        
        for u,v in edges:
            adj_graph[u].append(v)
            adj_graph[v].append(u)
            
        colors = [0] * V
        def is_safe(node,c):
            for adj_node in adj_graph[node]:
                if colors[adj_node] == c:
                    return False
            return True
            
        def dfs(node):
            if node == V:
                return True
            for c in range(1,m+1):
                if is_safe(node, c):
                    colors[node] = c
                    if dfs(node+1):
                        return True
                    colors[node] = 0
            return False
        return dfs(0)
```

##### 5. Flower Planting With No Adjacent

```python
class Solution:
    def gardenNoAdj(self, n: int, paths: List[List[int]]) -> List[int]:
        adj_graph = [[] for _ in range(n+1)]
        for u,v in paths:
            adj_graph[u].append(v)
            adj_graph[v].append(u)

        colors = [0] * (n+1)

        def is_safe(node,c):
            for adj_node in adj_graph[node]:
                if colors[adj_node] == c:
                    return False
            return True
    
        def dfs(node):
            if node > n:
                return True
            for c in range(1,5):
                if is_safe(node,c):
                    colors[node] = c
                    if dfs(node+1):
                        return True
                    colors[node] = 0
            return False

        dfs(1)
        return colors[1:]
```

##### 1. Topological Sort using DFS

```python
# Topological Sort - Linear ordering of vertices such that if there is 
# an edge between u and v appears before v in that ordering

# Approach 1 -> Use DFS: Store the nodes whose DFS traversal completes first
from collections import deque
class Solution:
    def topoSort(self, V, adj_graph):
        vis = [False] * V
        q = deque()
        def dfs(node):
            vis[node] = True
            for adj_node in adj_graph[node]:
                if not vis[adj_node]:
                    dfs(adj_node)
            q.appendleft(node)
        
        for i in range(V):
            if not vis[i]:
                dfs(i)
        return list(q)
    
```

##### 18.  Find Eventual Safe States

```python
class Solution:
    def eventualSafeNodes(self, graph: List[List[int]]) -> List[int]:
        # nodes in a cycle can not reach any terminal node
        n = len(graph)
        vis = [0 for i in range(n)]
        # detect cycle in DFS
        def dfs(i):
            vis[i] = 2
            for adj_node in graph[i]:
                if vis[adj_node] == 0:
                    if dfs(adj_node):
                        return True
                elif vis[adj_node] == 2:
                    return True
            vis[i] = 1
            return False
        for i in range(n):
            if vis[i] == 0:
                dfs(i)
        # trick is that if vis[node] = 2, that means the node is in 
        # a cycle and never reaches the line vis[node] = 1. 
        safe_nodes = []
        for i in range(n):
            if vis[i] == 1:
                safe_nodes.append(i)
        return safe_nodes
```

##### 19. Dijkstra's Algorithm

```python
from typing import List
import queue

def dijkstra(edge: List[List[int]], vertices: int, edges: int, source: int):
    # Write your code here.
    # edge' contains {u, v, distance} vectors.
    graph = [[] for i in range(0,vertices)]
    dist  = [1e9] * vertices
    for (u,v,d) in edge:
        graph[u].append((v,d))
        graph[v].append((u,d))
    pq = queue.PriorityQueue()
    pq.put((0,source))
    dist[source] = 0
    
    while len(pq.queue) != 0:
        d,node = pq.get()
        for adj_node,weight in graph[node]:
            if dist[adj_node] > d + weight:
                dist[adj_node] = d + weight
                pq.put((dist[adj_node],adj_node))
        
    return dist
    
```

##### 20. 1091. Shortest Path in Binary Matrix

```python
import queue
class Solution:
    def shortestPathBinaryMatrix(self, grid) -> int:
        n = len(grid)
        if (grid[0][0] == 1 or grid[n-1][n-1]==1):
            return -1
        if n==1 and grid[0][0] == 0:
            return 1

        vis = [[False for i in range(n)] for j in range(n)]
        vis[0][0] = True
        q = queue.Queue()
        q.put((0,0,1))
        dr = [-1,-1,-1,1,1,1,0,0]
        dc = [-1,0,1,-1,0,1,-1,1]

        while len(q.queue) != 0:
            r,c,d = q.get()
            for k in range(8):
                new_r,new_c = r + dr[k], c + dc[k]
                if new_r >=0 and new_r < n and new_c >=0 and new_c < n and not vis[new_r][new_c] and not grid[new_r][new_c]:
                    if new_r==n-1 and new_c==n-1:
                        return d+1
                    vis[new_r][new_c] = True
                    q.put((new_r,new_c,d+1))
        return -1
```

##### 21. 1631. Path With Minimum Effort

```python
import queue
class Solution:
    def minimumEffortPath(self, heights: List[List[int]]) -> int:
        m,n = len(heights), len(heights[0])
        dr,dc = [0,0,1,-1],[1,-1,0,0]
        pq = queue.PriorityQueue()
        pq.put((0,0,0))
        dist = [[1e9 for j in range(n)] for i in range(m)]
        while len(pq.queue) != 0:
            diff,r,c = pq.get()
            if r==m-1 and c==n-1:
                return diff
            for k in range(0,4):
                new_r,new_c = r + dr[k], c + dc[k]
                if new_r >= 0 and new_r < m and new_c >=0 and new_c < n:
                    new_effort = max(abs(heights[new_r][new_c]-heights[r][c]),diff)
                    if new_effort < dist[new_r][new_c]:
                        dist[new_r][new_c] = new_effort
                        pq.put((new_effort,new_r,new_c))
        return -1
```

##### 22. 787. Cheapest Flights Within K Stops

```python
import queue
class Solution:
    def findCheapestPrice(self, n: int, flights: List[List[int]], src: int, dst: int, k: int) -> int:
        dist = [1e9] * n
        graph = [[] for i in range(n)]
        for u,v,w in flights:
            graph[u].append((v,w))
        pq = queue.PriorityQueue()
        pq.put((0,src,0))
        while len(pq.queue) != 0:
            stops,node,cost = pq.get()
            for adj_node,weight in graph[node]:
                if dist[adj_node] > cost + weight and stops <= k:
                    dist[adj_node] = cost + weight
                    pq.put((stops+1,adj_node,dist[adj_node]))
        if dist[dst] >= 1e9:
            return -1
        return dist[dst]
```

```python
import queue
class Solution:
    def countPaths(self, n: int, roads: List[List[int]]) -> int:
        # we need to consider the ways with the same dist
        graph = [[] for i in range(n)]
        for u,v,w in roads:
            graph[u].append((v,w))
            graph[v].append((u,w))

        dist = [1e12] * n
        ways = [0] * n
        pq = queue.PriorityQueue()
        pq.put((0,0))   # dist, node
        dist[0] = 0
        ways[0] = 1
        while len(pq.queue) != 0:
            d,node = pq.get()
            for adj_node,weight in graph[node]:
                if dist[adj_node] > d + weight:
                    ways[adj_node] = ways[node]
                    dist[adj_node] = d + weight
                    pq.put((dist[adj_node],adj_node))
                elif dist[adj_node] == d + weight:
                    ways[adj_node] = (ways[adj_node] + ways[node]) % (1e9+7)
        return int(ways[n-1])
```
