##### 1. BFS and 2. DFS

```python
class Solution:
    def dfsOfGraph(self, V, adj):
        dfs_order = []
        vis = [False] * V

        def dfs(node):
          vis[node] = True
          dfs_order.append(node)
          for adj_node in adj[node]:
            if not vis[adj_node]:
              dfs(adj_node)
          
        for v in range(V):
          if not vis[v]:
            dfs(v)
        return dfs_order

    
    def bfsOfGraph(self, V, adj):
      bfs_order = []
      vis = [False] * V

      def bfs(start_node):
        q = deque()
        q.append(start_node)
        vis[start_node] = True

        while len(q):
          node = q.popleft()
          bfs_order.append(node)
          for adj_node in adj[node]:
            if not vis[adj_node]:
              q.append(adj_node)
              vis[adj_node] = True
      for v in range(V):
        if not vis[v]:
          bfs(v)
      return bfs_order
```

##### 3. No. of provinces

```python
def findCircleNum(isConnected):
    # use BFS
    nodes,ans = len(isConnected),0
    vis = [False] * nodes
    def bfs(start):
        q = queue.Queue()
        q.put(start)
        while len(q.queue) != 0:
            curr = q.get()
            for j in range(nodes):
                if curr != j and isConnected[curr][j] and vis[j] == False:
                    q.put(j)
                    vis[j] = True
    for i in range(nodes):
        if vis[i] == False:
            ans += 1
            bfs(i)
    return ans

findCircleNum([[1,1,0],[1,1,0],[0,0,1]])
```

##### 4. Rotten Oranges

```python

def orangesRotting( grid: List[List[int]]) -> int:
    # Multi source BFS
    m, n = len(grid), len(grid[0])
    q = deque()
    fresh = 0

    for i in range(m):
        for j in range(n):
            if grid[i][j] == 2:
                q.append((i, j, 0))  # store (x, y, time)
            elif grid[i][j] == 1:
                fresh += 1

    # Step 2: BFS
    directions = [[0,1],[1,0],[0,-1],[-1,0]]
    minutes = 0

    while q:
        x, y, time = q.popleft()
        for dx, dy in directions:
            nx, ny = x + dx, y + dy
            if 0 <= nx < m and 0 <= ny < n and grid[nx][ny] == 1:
                grid[nx][ny] = 2  # rot the fresh orange
                fresh -= 1
                q.append((nx, ny, time + 1))
                minutes = time + 1

    return minutes if fresh == 0 else -1

orangesRotting([[2,1,1],[1,1,1],[0,1,2]])
```

##### 5. Flood Fill

```python
class Solution:
    def floodFill(self, grid: List[List[int]], sr: int, sc: int, color: int) -> List[List[int]]:
        m, n = len(grid), len(grid[0])

        q = deque()
        og_color = grid[sr][sc]
        if og_color == color:
            return grid

        grid[sr][sc] = color
        q.append((sr,sc))

        dij = [[-1,0],[1,0],[0,-1],[0,1]]

        while q:
            x,y = q.popleft()
            for dx,dy in dij:
                new_x, new_y = x+dx, y+dy
                if 0 <= new_x < m and 0 <= new_y < n and grid[new_x][new_y] == og_color:
                    q.append((new_x,new_y))
                    grid[new_x][new_y] = color
                
        return grid
```

##### 6. Cycle Detection in Undirected graph

```python
def isCycle(V: int, adj) -> bool:
    n = len(adj)
    vis = [False for j in range(n)]

    def bfs(src):
        q = queue.Queue()
        q.put((src,-1))
        vis[src] = True

        while len(q.queue) != 0:
            node,parent = q.get()
            for adj_node in adj[node]:
                if vis[adj_node] == False:
                    vis[adj_node] = True
                    q.put((adj_node,node))
                elif vis[adj_node] == True and parent != adj_node:
                    return True
        return False

    # disconnected components
    for i in range(n):
        if vis[i] == False:
            if bfs(i) == True:
                return True
    return False  
```

##### 14. Detecting a cycle in Directed graph

```python
def isCyclic(self, V, adj_graph):
    ''' DFS to find the cycle in Directed graph'''
    
    def dfs(i):
        ''' DFS helper function to check the cycle'''
        vis[i] = True
        path_vis[i] = True
        for adj_node in adj_graph[i]:
            if vis[adj_node] == False:
                if dfs(adj_node):
                    return True
            elif path_vis[adj_node]: 
                # node is already visited in the path
                return True
        path_vis[i] = False
        return False
        
    vis = [False for i in range(V)]
    path_vis = [False for i in range(V)]
    
    for i in range(V):
        if vis[i] == 0:
            if dfs(i):
                return True
    return False
```

```python

def updateMatrix(mat: List[List[int]]) -> List[List[int]]:
    m,n = len(mat), len(mat[0])

    vis = [[False for _ in range(n)] for _ in range(m)]
    ans = [[0 for _ in range(n)] for _ in range(m)]
    q = deque()

    # multi source bfs

    for i in range(m):
        for j in range(n):
            if mat[i][j] == 0:
                q.append((i,j,0))
                vis[i][j] = True

    dxy = [[-1,0],[1,0],[0,-1],[0,1]]
    while q:
        x,y,d = q.popleft()
        ans[x][y] = d
        for dx,dy in dxy:
            new_x, new_y = x + dx, y + dy
            if 0 <= new_x < m and 0 <= new_y < n and not vis[new_x][new_y]:
                q.append((new_x,new_y,d+1))
                vis[new_x][new_y] = 1
    return ans


```

##### 8. Surrounded Regions

```python
class Solution:
    def solve(self, board) -> None:
        """
        Do not return anything, modify board in-place instead.
        """
        di,dj = [0,0,1,-1],[1,-1,0,0]
        m, n = len(board), len(board[0])
        vis = [[False for j in range(n)] for i in range(m)]
        def dfs(i,j,m,n):
            vis[i][j] = True
            for k in range(4):
                new_i,new_j = i+di[k],j+dj[k]
                if new_i>=0 and new_i<m and new_j>=0 and new_j<n and board[new_i][new_j]=='O' and vis[new_i][new_j] == False:
                    dfs(new_i,new_j,m,n)
        # column wise 
        for j in range(n):
            # first row
            if board[0][j] == 'O' and vis[0][j] == False:
                dfs(0,j,m,n)
            # last row
            if board[m-1][j] == 'O' and vis[m-1][j] == False:
                dfs(m-1,j,m,n)
        # row wise
        for i in range(m):
            # first column
            if board[i][0] == 'O' and vis[i][0] == False:
                dfs(i,0,m,n)
            # last column
            if board[i][n-1] == 'O' and vis[i][n-1] == False:
                dfs(i,n-1,m,n)
            
        for i in range(m):
            for j in range(n):
                if board[i][j] == 'O' and vis[i][j] == False:
                    board[i][j] = 'X'

        
```

##### 9. No. of enclaves - Flood fill variation

```python
def numEnclaves(grid) -> int:
    m,n = len(grid), len(grid[0])
    di,dj = [0,0,1,-1],[-1,1,0,0]

    def dfs(i,j,m,n):
        grid[i][j] = 0
        for k in range(4):
            new_i,new_j=i+di[k],j+dj[k]
            if new_i>=0 and new_i<m and new_j>=0 and new_j<n and grid[new_i][new_j]:
                dfs(new_i,new_j,m,n)

    # column wise
    for i in range(m):
        # 1st column
        if grid[i][0]:
            dfs(i,0,m,n)
        # last column
        if grid[i][n-1]:
            dfs(i,n-1,m,n)
    # row wise
    for j in range(n):
        # 1st row
        if grid[0][j]:
            dfs(0,j,m,n)
        # last row
        if grid[m-1][j]:
            dfs(m-1,j,m,n)

    ans = 0
    for i in range(m):
        for j in range(n):
            if grid[i][j]:
                ans+=1
    return ans
```

##### 12. No. of Distinct Islands

```python
def countDistinctIslands(grid) -> int:
    m,n = len(grid), len(grid[0])
    vis = [[False for i in range(n)] for j in range(m)]
    st = set()
    di,dj = [0,0,1,-1],[1,-1,0,0]

    def dfs(i,j,m,n,temp,bi,bj):
        vis[i][j] = True
        temp.append((i-bi,j-bj))
        for k in range(4):
            new_i,new_j = i+di[k],j+dj[k]
            if new_i>=0 and new_i<m and new_j>=0 and new_j<n and grid[new_i][new_j] and vis[new_i][new_j]==False:
                dfs(new_i,new_j,m,n,temp,bi,bj)

    for i in range(m):
        for j in range(n):
            if vis[i][j] == False and grid[i][j]:
                temp = []
                dfs(i,j,m,n,temp,i,j)
                st.add(tuple(temp))
    return len(st)
```

##### 13. Bipartite Graph

```python
import queue
def isBipartite(self, graph) -> bool:
    n = len(graph)
    color = [-1 for j in range(n)]

    def bfs_coloring(src,n):
        q = queue.Queue()
        q.put((src,0))  # node, color
        color[src] = 0
        while len(q.queue) != 0:
            node, col = q.get()
            for adj_node in graph[node]:
                if color[adj_node]==-1:
                    q.put((adj_node, not col))
                    vis[adj_node] = True
                    color[adj_node] = not col
                elif color[adj_node] == col:
                    return False
        return True
    # Edge case - disconnected components
    for i in range(n):
        if color[i] == -1:
            if bfs_coloring(i,n) == False:
                return False
    return True
```

##### 16. Course Schedule I

```python
def canFinish(self, numCourses: int, prerequisites: List[List[int]]) -> bool:
    # prerequisites can be taken as (u,v) edge
    # if there is no cycle in the graph, courses can be finished. 

    graph = [[] for i in range(numCourses)]
    for u,v in prerequisites:
        graph[u].append(v)
    vis = [0 for i in range(numCourses)]

    # detect cycle in DFS
    def dfs(i):
        vis[i] = 2
        for adj_node in graph[i]:
            if vis[adj_node] == 0:
                if dfs(adj_node):
                    return True
            elif vis[adj_node] == 2:
                return True
        vis[i] = 1
        return False

    for i in range(numCourses):
        if vis[i] == False:
            if dfs(i):
                return False
    return True
```

```python
def findOrder(self, numCourses: int, prerequisites: List[List[int]]) -> List[int]:
    graph = [[] for i in range(numCourses)]
    for u,v in prerequisites:
        graph[u].append(v)
    vis = [0 for i in range(numCourses)]
    topo_sort = []

    # detect cycle in DFS
    def dfs(i):
        vis[i] = 2
        for adj_node in graph[i]:
            if vis[adj_node] == 0:
                if dfs(adj_node):
                    return True
            elif vis[adj_node] == 2:
                return True
        vis[i] = 1
        topo_sort.append(i)
        return False
        
    for i in range(numCourses):
        if vis[i] == False:
            if dfs(i):
                return []
    return topo_sort
```

##### 18.  Find Eventual Safe States

```python
# approach 1

def eventualSafeNodes(self, graph: List[List[int]]) -> List[int]:
    # nodes in a cycle can not reach any terminal node
    n = len(graph)
    vis = [0 for i in range(n)]
    # detect cycle in DFS
    def dfs(i):
        vis[i] = 2
        for adj_node in graph[i]:
            if vis[adj_node] == 0:
                if dfs(adj_node):
                    return True
            elif vis[adj_node] == 2:
                return True
        vis[i] = 1
        return False
    for i in range(n):
        if vis[i] == 0:
            dfs(i)
    # trick is that if vis[node] = 2, that means the node is in 
    # a cycle and never reaches the line vis[node] = 1. 
    safe_nodes = []
    for i in range(n):
        if vis[i] == 1:
            safe_nodes.append(i)
    return safe_nodes
```

##### 19. Dijkstra's Algorithm

```python
from typing import List
import queue

def dijkstra(edge: List[List[int]], vertices: int, edges: int, source: int):
    # Write your code here.
    # edge' contains {u, v, distance} vectors.
    graph = [[] for i in range(0,vertices)]
    dist  = [1e9] * vertices
    for (u,v,d) in edge:
        graph[u].append((v,d))
        graph[v].append((u,d))
    pq = queue.PriorityQueue()
    pq.put((0,source))
    dist[source] = 0
    
    while len(pq.queue) != 0:
        d,node = pq.get()
        for adj_node,weight in graph[node]:
            if dist[adj_node] > d + weight:
                dist[adj_node] = d + weight
                pq.put((dist[adj_node],adj_node))
        
    return dist
    
```

##### 20. 1091. Shortest Path in Binary Matrix

```python
import queue
class Solution:
    def shortestPathBinaryMatrix(self, grid) -> int:
        n = len(grid)
        if (grid[0][0] == 1 or grid[n-1][n-1]==1):
            return -1
        if n==1 and grid[0][0] == 0:
            return 1

        vis = [[False for i in range(n)] for j in range(n)]
        vis[0][0] = True
        q = queue.Queue()
        q.put((0,0,1))
        dr = [-1,-1,-1,1,1,1,0,0]
        dc = [-1,0,1,-1,0,1,-1,1]

        while len(q.queue) != 0:
            r,c,d = q.get()
            for k in range(8):
                new_r,new_c = r + dr[k], c + dc[k]
                if new_r >=0 and new_r < n and new_c >=0 and new_c < n and not vis[new_r][new_c] and not grid[new_r][new_c]:
                    if new_r==n-1 and new_c==n-1:
                        return d+1
                    vis[new_r][new_c] = True
                    q.put((new_r,new_c,d+1))
        return -1
```

##### 21. 1631. Path With Minimum Effort

```python
import queue
class Solution:
    def minimumEffortPath(self, heights: List[List[int]]) -> int:
        m,n = len(heights), len(heights[0])
        dr,dc = [0,0,1,-1],[1,-1,0,0]
        pq = queue.PriorityQueue()
        pq.put((0,0,0))
        dist = [[1e9 for j in range(n)] for i in range(m)]
        while len(pq.queue) != 0:
            diff,r,c = pq.get()
            if r==m-1 and c==n-1:
                return diff
            for k in range(0,4):
                new_r,new_c = r + dr[k], c + dc[k]
                if new_r >= 0 and new_r < m and new_c >=0 and new_c < n:
                    new_effort = max(abs(heights[new_r][new_c]-heights[r][c]),diff)
                    if new_effort < dist[new_r][new_c]:
                        dist[new_r][new_c] = new_effort
                        pq.put((new_effort,new_r,new_c))
        return -1
```

##### 22. 787. Cheapest Flights Within K Stops

```python
import queue
class Solution:
    def findCheapestPrice(self, n: int, flights: List[List[int]], src: int, dst: int, k: int) -> int:
        dist = [1e9] * n
        graph = [[] for i in range(n)]
        for u,v,w in flights:
            graph[u].append((v,w))
        pq = queue.PriorityQueue()
        pq.put((0,src,0))
        while len(pq.queue) != 0:
            stops,node,cost = pq.get()
            for adj_node,weight in graph[node]:
                if dist[adj_node] > cost + weight and stops <= k:
                    dist[adj_node] = cost + weight
                    pq.put((stops+1,adj_node,dist[adj_node]))
        if dist[dst] >= 1e9:
            return -1
        return dist[dst]
```

##### 24. 1976. Number of Ways to Arrive at Destination

```python
import queue
class Solution:
    def countPaths(self, n: int, roads: List[List[int]]) -> int:
        # we need to consider the ways with the same dist
        graph = [[] for i in range(n)]
        for u,v,w in roads:
            graph[u].append((v,w))
            graph[v].append((u,w))

        dist = [1e12] * n
        ways = [0] * n
        pq = queue.PriorityQueue()
        pq.put((0,0))   # dist, node
        dist[0] = 0
        ways[0] = 1
        while len(pq.queue) != 0:
            d,node = pq.get()
            for adj_node,weight in graph[node]:
                if dist[adj_node] > d + weight:
                    ways[adj_node] = ways[node]
                    dist[adj_node] = d + weight
                    pq.put((dist[adj_node],adj_node))
                elif dist[adj_node] == d + weight:
                    ways[adj_node] = (ways[adj_node] + ways[node]) % (1e9+7)
        return int(ways[n-1])
```
