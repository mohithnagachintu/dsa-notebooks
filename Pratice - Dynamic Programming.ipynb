### 1D DP 

##### 1. Climbing Stars

```python
def climbStairs(n):
    dp = [-1] * n

    def solve_rec(i,n):
        # choice - take either 1 or 2 steps 
        if i==n-1: 
            return 1
        elif i==n-2:
            return 2
        return solve_rec(i+1,n) + solve_rec(i+2,n)

    def solve_memo(i,n):
        if i == n-1:
            return 1
        elif i == n-2:
            return 2
        elif dp[i] != -1:
            return dp[i]
        dp[i] = solve_memo(i+1,n) + solve_memo(i+2,n)
        return dp[i]
    
    def solve_tab(n):
        if n==1:
            return 1
        elif n==2:
            return 2
        else:
            dp = [0] * n
            dp[0],dp[1] = 1,2
            for i in range(2,n):
                dp[i] = dp[i-1] + dp[i-2]
        return dp[n-1]
    return solve_tab(n)
```

##### 2. Frog Jump

```python
def frogJump(n: int, heights) :
    # When a frog is at ith case, it can jump to i+1th or i+2th 
    dp = [0] * (n)
    dp[1] = abs(heights[1]-heights[0])
    for i in range(2,n):
        dp[i] = min(dp[i-1] + abs(heights[i]-heights[i-1]),dp[i-2] + abs(heights[i]-heights[i-2]))
    return dp[n-1]
frogJump(4,[10,20,30,10])
```

##### 3. Frog Jump with k distances

```python
def minimizeCost(n : int, k : int, heights : List[int]) -> int:
    if n==1:
        return 0
    dp = [0] * n
    dp[1] = abs(heights[1] - heights[0])
    for i in range(2,n):
        temp = 1e9
        for j in range(max(0,i-k),i):
            temp = min(temp,dp[j] + abs(heights[i]-heights[j]))
        dp[i] = temp
    return dp[n-1]
```

##### 4.Maximum sum of non-adjacent elements 

```python
def rob(nums):
    n = len(nums)
    if n==1: 
        return nums[0]
    elif n==2:
        return max(nums[0],nums[1])
    elif n==3:
        return max(nums[1],nums[0] + nums[2])
    else:
        dp = [0] * (n+1)
        dp[0] = nums[0]
        dp[1] = nums[1]
        dp[2] = nums[0] + nums[2]
        for i in range(3,n):
            dp[i] = nums[i] + max(dp[i-2],dp[i-3])
        dp[n] = max(dp[n-1],dp[n-2])
        return dp[n]
    return -1
```

##### 5. House Robber

```python
class Solution:
    # house robber 1 solution
    def robber1(self,nums,start,end):
        n = end - start + 1
        if n==1: 
            return nums[start]
        elif n==2:
            return max(nums[start],nums[start + 1])
        elif n==3:
            return max(nums[start + 1],nums[start] + nums[start + 2])
        else:
            dp = [0] * (n+1)
            dp[0] = nums[start]
            dp[1] = nums[start + 1]
            dp[2] = nums[start] + nums[start + 2]
            for i in range(3,n):
                dp[i] = nums[start + i] + max(dp[i-2],dp[i-3])
            dp[n] = max(dp[n-1],dp[n-2])
            return dp[n]
        return -1

    def rob(self, nums) -> int:
        n = len(nums)
        if n==1: return nums[0]
        # two linear robber patterns -> [0...n-2] or [1...n-1]
        return max(self.robber1(nums,0,n-2),self.robber1(nums,1,n-1))
```

##### 6. Ninja's Training (DP 7)

```python
from typing import *

  
def ninjaTraining(n: int, points: List[List[int]]) -> int:
    dp = [[-1 for j in range(4)] for i in range(n)]
    def solve(i,previous):
        if dp[i][previous] != -1:
            return dp[i][previous]
        if i == 0: 
            maxi = 0
            for j in range(3):
                if j!=previous:
                    maxi = max(maxi,points[0][j])
            dp[0][previous] = maxi
            return maxi

        maxi = 0
        for j in range(3):
            if j != previous:
                pts = points[i][j]  + solve(i-1,j)
                maxi = max(maxi,pts)
        dp[i][previous] = maxi
        return maxi
    return solve(n-1,3)
```

##### 7. Grid Unique Paths 

```python
def uniquePaths(self, m: int, n: int) -> int:
    dp = [[1 for i in range(n)] for i in range(m)]
    for i in range(1,m):
        for j in range(1,n):
            dp[i][j] = dp[i][j-1] + dp[i-1][j]
    return dp[-1][-1]
```

##### 8. Grid Unique Paths 2

```python
def uniquePathsWithObstacles(self, grid: List[List[int]]) -> int:
    n,m = len(grid),len(grid[0])
    dp = [[0 for j in range(m)] for i in range(n)]
    for j in range(m):
        if grid[0][j]: break
        dp[0][j] = 1
    for i in range(n):
        if grid[i][0]: break
        dp[i][0] = 1
    for i in range(1,n):
        for j in range(1,m):
            if not grid[i][j]: 
                dp[i][j] = dp[i-1][j] + dp[i][j-1]
    return dp[-1][-1]
```

##### 9. Minimum Path Sum In a Grid

```python
def minPathSum(self, grid: List[List[int]]) -> int:
    n,m = len(grid),len(grid[0])
    dp = [[0 for j in range(m)] for i in range(n)]
    dp[0][0] = grid[0][0]
    for j in range(1,m):
        dp[0][j] = dp[0][j-1] + grid[0][j]
    for i in range(1,n):
        dp[i][0] = dp[i-1][0] + grid[i][0]
    for i in range(1,n):
        for j in range(1,m):
            dp[i][j] = grid[i][j] + min(dp[i-1][j],dp[i][j-1])
    return dp[-1][-1]
```

##### 10. Minimum path sum in Triangular Grid

```python
def minimumTotal(self, triangle: List[List[int]]) -> int:
    n = len(triangle)
    for i in range(n-2,-1,-1):
        for j in range(i+1):
            triangle[i][j] += min(triangle[i+1][j],triangle[i+1][j+1])
    return triangle[0][0]
```

##### 11. Minimum/Maximum Falling Path Sum

```python
def minFallingPathSum(self, matrix: List[List[int]]) -> int:
    n = len(matrix)
    dp = [[0 for i in range(n)] for j in range(n)]
    for i in range(n):
        dp[0][i] = matrix[0][i]
    
    for i in range(1,n):
        for j in range(n):
            temp = dp[i-1][j]
            if j-1>=0:
                temp = min(temp,dp[i-1][j-1])
            if j+1<n:
                temp = min(temp,dp[i-1][j+1])
            dp[i][j] = matrix[i][j] + temp

    return min(dp[-1])
```

##### 12. 3-d DP : Ninja and his friends

```python
def maximumChocolates(n: int, m: int, grid: List[List[int]]) -> int:
    # i,j and i,k
    # dp[i][j][k]
    dp = [[[-1 for k in range(m)] for j in range(m)] for i in range(n)]
    def solve(n,m,i,j,k):
        if j < 0 or k < 0 or i == n or j == m or k==m: 
            return -1e8
        if dp[i][j][k] != -1:
            return dp[i][j][k]

        if i==n-1:
            if j==k:return grid[i][j]
            else: return grid[i][j] + grid[i][k]
        d  = [-1,0,1]
        ans = -1e8
        for x in d:
            for y in d:
                temp = 0
                if j == k:
                    temp = grid[i][j]
                else:
                    temp = grid[i][j] + grid[i][k]
                temp += solve(n,m,i+1,j+x,k+y)
                ans = max(ans,temp)
        dp[i][j][k] = ans
        return ans

    return solve(n,m,0,0,m-1)
```

##### 17. Count Partitions with Given Difference

```python
class Solution:
    def count_subsets(self, arr, target):
        dp = [[-1 for _ in range(target+1)] for _ in range(len(arr))]
        
        def solve(i, target):
            # base case
            if i == len(arr) - 1:
                if target == 0:
                    if arr[i] == 0:
                        return 2   # pick and dont_pick 
                    else:
                        return 1
                elif target == arr[i]:
                    return 1
                return 0
            
            if dp[i][target] != -1:
                return dp[i][target]
                    
            dont_pick = solve(i+1, target)
            pick = 0
            if arr[i] <= target:
                pick = solve(i+1, target-arr[i])
            dp[i][target] = pick + dont_pick
            return dp[i][target]
            
        return solve(0, target)    
        
    def countPartitions(self, arr, d):
        # s1 - s2 = d
        # s1 + s2 = t
        # s1 = (d + t) // 2
        s = sum(arr)
        if (d+s) & 1:
            return 0
        s1 = (d + s)//2
        return self.count_subsets(arr, s1)
        
```

##### 20. Target Sum

```python
class Solution:
    def count_subset_sum(self, nums, target):
        n = len(nums)
        dp = [[0 for j in range(target+1)] for i in range(n)]
        for j in range(target+1):
            if j == 0:
                if nums[n-1] == 0:
                    dp[n-1][j] = 2
                else:
                    dp[n-1][j] = 1
            elif nums[n-1] == j:
                dp[n-1][j] = 1
        
        for i in range(n-2,-1,-1):
            for j in range(target+1):
                dont_pick = dp[i+1][j]
                pick = 0
                if nums[i] <= j:
                    pick = dp[i+1][j-nums[i]]
                dp[i][j] = dont_pick + pick 
        return dp[0][target]

    def findTargetSumWays(self, nums: List[int], target: int) -> int:
        # s1 + s2 = sum(nums) = s
        # s1 - s2 = target 
        # s1 = (s - s2) // 2
        n, s = len(nums), sum(nums) 
        if (s - target) & 1 or (s - target) < 0:
            return 0
        return self.count_subset_sum(nums, (s-target) // 2)
```

##### 638. Shopping Offers

```python
class Solution:
    def shoppingOffers(self, price: List[int], special: List[List[int]], needs: List[int]) -> int:
        n,m = len(price), len(special)
        mapper = dict()

        def solve(n,m,j):
            if j == m: 
                # take single thing price
                s = 0
                for i in range(n):
                    s += (needs[i] * price[i])
                return s
            
            # key for mapper
            key = ''.join(map(str, needs)) + str(j)
            if key in mapper:
                return mapper[key]


            # check if we can apply this special offer special[j]
            can_apply = True
            result = None
            for i in range(n):
                if needs[i] < special[j][i]:
                    can_apply = False
                    break

            if can_apply:
                # take this offer
                for i in range(n):
                    needs[i] -= special[j][i]
                a = special[j][-1] + solve(n,m,j)  # unbounded
                for i in range(n):
                    needs[i] += special[j][i]
                
                # don't take this offer
                b = solve(n,m,j+1)
                result = min(a,b)
            else:
                result =  solve(n,m,j+1)
            mapper[key] = result
            return result

        return solve(n,m,0)
```

### Longest Palindromic SubString - Very Important Approach

```python
def longestPalindrome(self, s: str) -> str:
    n = len(s)
    dp = [[False for _ in range(n)] for _ in range(n)]
    ans = [0,0]

    # odd length:
        # 1 - Always palindrome
        # 3 - check
        # 5 - check 
    # even length:
        # 2 - s[i] == s[i+1] - Palindrome

    # length - 1 initialization
    for i in range(n): 
        dp[i][i] = True 

    # length - 2 initialization        
    for i in range(n-1): 
        if s[i] == s[i+1]:
            dp[i][i+1] = True
            ans = [i,i+1]

    # substrs with len 3,4,5...
    for diff in range(2,n):
        for i in range(n-diff):
            j = i + diff
            if s[i] == s[j] and dp[i+1][j-1] == True:
                dp[i][j] = True
                ans = [i,j]
    
    return s[ans[0]:ans[1]+1]
```

##### 32. Wildcard Matching

```python
def isMatch(self, s: str, p: str) -> bool:
    
    m,n = len(s), len(p)
    dp = [[-1 for _ in range(n)] for _ in range(m)]
    def solve(i: int, j: int) -> int: 
        m,n = len(s), len(p)
        # base cases
        if i == m and j == n:   # matched everything
            return True

        if i == m and j < n:  # string is done, pattern is pending 
            for _j in range(j,n):
                if p[_j] != '*':
                    return False 
            return True

        if j == n and i < m:  # pattern is done, string is pending
            return False

        if dp[i][j] != -1:
            return dp[i][j]
            
        temp = False
        if s[i] == p[j] or p[j] == '?':   # Match single character
            temp = solve(i+1,j+1)
        
        if p[j] == '*':     # Match zero or more characters
            temp = solve(i+1,j) or solve(i,j+1)
        dp[i][j] = temp
        return temp
    
    return solve(0,0)
```

```python
def isMatch(s: str, p: str) -> bool:
    m,n = len(s), len(p)
    dp = [[-1 for _ in range(n)] for _ in range(m)]

    def solve(i,j):
        if i == m and j == n:  # both ptrs reached end
            return True
        
        if i == m and j < n:   # s is exhausted
            while j + 1 < n and p[j + 1] == '*':
                j += 2
            return j == n
        
        if i < m and j == n:   # p is exhausted
            return False

        if dp[i][j] != -1:
            return dp[i][j]
        
        temp = False
        
        if j+1 < n and p[j+1] == '*':
            if s[i] == p[j] or p[j] == '.':   # if chars matches
                temp = solve(i,j+2)     or   solve(i+1,j)      # 
                      # skip * at j + 1      use * 
            else:
                temp = solve(i,j+2)          # skip * at j + 1
        elif s[i] == p[j] or p[j] == '.':
            temp = solve(i+1,j+1)
        
        dp[i][j] = temp
        return temp

    return solve(0,0)
```

##### 41. Longest Divisible Subset

```python

def largestDivisibleSubset( nums: List[int]) -> List[int]:
    nums.sort()
    n = len(nums)
    dp = [1] * n
    backtrace = [i for i in range(n)]
    ans,idx = 1,0
    
    for i in range(n):
        for j in range(i):
            if nums[i] % nums[j] == 0 and dp[i] < dp[j] + 1:
                dp[i] = dp[j] + 1
                backtrace[i] = j
            if dp[i] > ans:
                ans, idx = dp[i], i

    temp = [nums[idx]]
    while idx != backtrace[idx]:
        idx = backtrace[idx]
        temp.append(nums[idx])
    return temp
```

##### 42. Longest String Chain

```python

def longestStrChain( words: List[str]) -> int:
    words.sort(key = len)

    n = len(words)
    dp = [1] * n

    def compare(wa, wb):
        len_a, len_b = len(wa), len(wb)
        if len_a != len_b + 1: return False 
        i, j = 0,0
        while i < len_a:
            if j < len_b and wa[i] == wb[j]:
                i += 1
                j += 1
            else:
                i += 1
        return i == len_a and j == len_b

    for i in range(n):
        for j in range(i):
            if compare(words[i], words[j]) and dp[i] < dp[j] + 1:
                dp[i] = dp[j] + 1
    
    return max(dp)
        
```

```python
def countWays(s):
    n = len(s)
    dp = [[[-1 for _ in range(2)] for _ in range(n)] for _ in range(n)]    
    def solve(i,j,flag):
        if i > j:
            return 0
        if i == j:
            if s[i] == 'T' and flag: return 1
            if s[i] == 'F' and not flag: return 1
            return 0
        if dp[i][j][flag] != -1:
            return dp[i][j][flag]
        ways = 0
        for k in range(i+1,j,2):
            left_true = solve(i,k-1,True)
            left_false = solve(i,k-1,False)
            right_true = solve(k+1,j,True)
            right_false = solve(k+1,j,False)
            
            if s[k] == '&':
                if flag: # looking for True
                    ways += (left_true * right_true)
                else:    # looking for False
                    ways += (left_false * right_false) +\
                            (left_true * right_false) +\
                            (left_false * right_true) 
            elif s[k] == '^':
                if flag: # looking for True
                    ways += (left_false * right_true) +\
                            (left_true * right_false)
                else:    # looking for False
                    ways += (left_true * right_true) +\
                            (left_false * right_false)
            elif s[k] == '|':
                if flag: # looking for True
                    ways += (left_true * right_true) +\
                            (left_true * right_false) +\
                            (left_false * right_true)
                else:    # looking for False
                    ways += (left_false * right_false)
        dp[i][j][flag] = ways
        return ways
    return solve(0,n-1,True)

countWays('T|T&F^T')
```

##### 357. Count Numbers with Unique Digits

```python
def countNumbersWithUniqueDigits(n: int) -> int:
    if n==0: return 1
    if n==1: return 10
    def solve(n):
        ans,c = 9, 9
        for i in range(1,n):
            ans *= c
            c -= 1
        return ans
    ans = 10
    for j in range(2,n+1):
        ans += solve(j)
    return ans
```
