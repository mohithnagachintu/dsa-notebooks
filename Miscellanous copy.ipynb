##### 2. Kth Smallest Lexicographical Number

```python
class Solution:
    def findKthNumber(self, n: int, k: int) -> int:
        cur = 1
        i = 1   # index in the lexicographical order (1-based)

        def count(cur):
            steps = 0
            neighbor = cur + 1
            while cur <= n:
                temp = min(neighbor,n+1)   # take numbers only less than equal to n
                steps += temp - cur
                neighbor *= 10
                cur *= 10
            return steps


        while i < k:
            steps = count(cur)
            if i + steps <= k:  # skip this tree
                cur += 1
                i += steps 
            else:   # in this tree
                cur *= 10       # go to left child
                i += 1          

        return cur
```
