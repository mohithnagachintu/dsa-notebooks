### 1D DP 

##### 1. Climbing Stars

```python
def climbStairs(n):
    dp = [-1] * n

    def solve_rec(i,n):
        # choice - take either 1 or 2 steps 
        if i==n-1: 
            return 1
        elif i==n-2:
            return 2
        return solve_rec(i+1,n) + solve_rec(i+2,n)

    def solve_memo(i,n):
        if i == n-1:
            return 1
        elif i == n-2:
            return 2
        elif dp[i] != -1:
            return dp[i]
        dp[i] = solve_memo(i+1,n) + solve_memo(i+2,n)
        return dp[i]
    
    def solve_tab(n):
        if n==1:
            return 1
        elif n==2:
            return 2
        else:
            dp = [0] * n
            dp[0],dp[1] = 1,2
            for i in range(2,n):
                dp[i] = dp[i-1] + dp[i-2]
        return dp[n-1]
    return solve_tab(n)
```

##### 2. Frog Jump

```python
def frogJump(n: int, heights) :
    # When a frog is at ith case, it can jump to i+1th or i+2th 
    dp = [0] * (n)
    dp[1] = abs(heights[1]-heights[0])
    for i in range(2,n):
        dp[i] = min(dp[i-1] + abs(heights[i]-heights[i-1]),dp[i-2] + abs(heights[i]-heights[i-2]))
    return dp[n-1]
frogJump(4,[10,20,30,10])
```

##### 3. Frog Jump with k distances

```python
def minimizeCost(n : int, k : int, heights : List[int]) -> int:
    if n==1:
        return 0
    dp = [0] * n
    dp[1] = abs(heights[1] - heights[0])
    for i in range(2,n):
        temp = 1e9
        for j in range(max(0,i-k),i):
            temp = min(temp,dp[j] + abs(heights[i]-heights[j]))
        dp[i] = temp
    return dp[n-1]
```

##### 4.Maximum sum of non-adjacent elements 

```python
def rob(nums):
    n = len(nums)
    if n==1: 
        return nums[0]
    elif n==2:
        return max(nums[0],nums[1])
    elif n==3:
        return max(nums[1],nums[0] + nums[2])
    else:
        dp = [0] * (n+1)
        dp[0] = nums[0]
        dp[1] = nums[1]
        dp[2] = nums[0] + nums[2]
        for i in range(3,n):
            dp[i] = nums[i] + max(dp[i-2],dp[i-3])
        dp[n] = max(dp[n-1],dp[n-2])
        return dp[n]
    return -1
```

##### 5. House Robber

```python
class Solution:
    # house robber 1 solution
    def robber1(self,nums,start,end):
        n = end - start + 1
        if n==1: 
            return nums[start]
        elif n==2:
            return max(nums[start],nums[start + 1])
        elif n==3:
            return max(nums[start + 1],nums[start] + nums[start + 2])
        else:
            dp = [0] * (n+1)
            dp[0] = nums[start]
            dp[1] = nums[start + 1]
            dp[2] = nums[start] + nums[start + 2]
            for i in range(3,n):
                dp[i] = nums[start + i] + max(dp[i-2],dp[i-3])
            dp[n] = max(dp[n-1],dp[n-2])
            return dp[n]
        return -1

    def rob(self, nums) -> int:
        n = len(nums)
        if n==1: return nums[0]
        # two linear robber patterns -> [0...n-2] or [1...n-1]
        return max(self.robber1(nums,0,n-2),self.robber1(nums,1,n-1))
```

##### 6. Ninja's Training (DP 7)

```python
from typing import *

  
def ninjaTraining(n: int, points: List[List[int]]) -> int:
    dp = [[-1 for j in range(4)] for i in range(n)]
    def solve(i,previous):
        if dp[i][previous] != -1:
            return dp[i][previous]
        if i == 0: 
            maxi = 0
            for j in range(3):
                if j!=previous:
                    maxi = max(maxi,points[0][j])
            dp[0][previous] = maxi
            return maxi

        maxi = 0
        for j in range(3):
            if j != previous:
                pts = points[i][j]  + solve(i-1,j)
                maxi = max(maxi,pts)
        dp[i][previous] = maxi
        return maxi
    return solve(n-1,3)
```

##### 7. Grid Unique Paths 

```python
def uniquePaths(self, m: int, n: int) -> int:
    dp = [[1 for i in range(n)] for i in range(m)]
    for i in range(1,m):
        for j in range(1,n):
            dp[i][j] = dp[i][j-1] + dp[i-1][j]
    return dp[-1][-1]
```

##### 8. Grid Unique Paths 2

```python
def uniquePathsWithObstacles(self, grid: List[List[int]]) -> int:
    n,m = len(grid),len(grid[0])
    dp = [[0 for j in range(m)] for i in range(n)]
    for j in range(m):
        if grid[0][j]: break
        dp[0][j] = 1
    for i in range(n):
        if grid[i][0]: break
        dp[i][0] = 1
    for i in range(1,n):
        for j in range(1,m):
            if not grid[i][j]: 
                dp[i][j] = dp[i-1][j] + dp[i][j-1]
    return dp[-1][-1]
```

##### 9. Minimum Path Sum In a Grid

```python
def minPathSum(self, grid: List[List[int]]) -> int:
    n,m = len(grid),len(grid[0])
    dp = [[0 for j in range(m)] for i in range(n)]
    dp[0][0] = grid[0][0]
    for j in range(1,m):
        dp[0][j] = dp[0][j-1] + grid[0][j]
    for i in range(1,n):
        dp[i][0] = dp[i-1][0] + grid[i][0]
    for i in range(1,n):
        for j in range(1,m):
            dp[i][j] = grid[i][j] + min(dp[i-1][j],dp[i][j-1])
    return dp[-1][-1]
```

##### 10. Minimum path sum in Triangular Grid

```python
def minimumTotal(self, triangle: List[List[int]]) -> int:
    n = len(triangle)
    for i in range(n-2,-1,-1):
        for j in range(i+1):
            triangle[i][j] += min(triangle[i+1][j],triangle[i+1][j+1])
    return triangle[0][0]
```

##### 11. Minimum/Maximum Falling Path Sum

```python
def minFallingPathSum(self, matrix: List[List[int]]) -> int:
    n = len(matrix)
    dp = [[0 for i in range(n)] for j in range(n)]
    for i in range(n):
        dp[0][i] = matrix[0][i]
    
    for i in range(1,n):
        for j in range(n):
            temp = dp[i-1][j]
            if j-1>=0:
                temp = min(temp,dp[i-1][j-1])
            if j+1<n:
                temp = min(temp,dp[i-1][j+1])
            dp[i][j] = matrix[i][j] + temp

    return min(dp[-1])
```

##### 12. 3-d DP : Ninja and his friends

```python
def maximumChocolates(n: int, m: int, grid: List[List[int]]) -> int:
    # i,j and i,k
    # dp[i][j][k]
    dp = [[[-1 for k in range(m)] for j in range(m)] for i in range(n)]
    def solve(n,m,i,j,k):
        if j < 0 or k < 0 or i == n or j == m or k==m: 
            return -1e8
        if dp[i][j][k] != -1:
            return dp[i][j][k]

        if i==n-1:
            if j==k:return grid[i][j]
            else: return grid[i][j] + grid[i][k]
        d  = [-1,0,1]
        ans = -1e8
        for x in d:
            for y in d:
                temp = 0
                if j == k:
                    temp = grid[i][j]
                else:
                    temp = grid[i][j] + grid[i][k]
                temp += solve(n,m,i+1,j+x,k+y)
                ans = max(ans,temp)
        dp[i][j][k] = ans
        return ans

    return solve(n,m,0,0,m-1)
```

##### 13. Subset sum equal to target

```python
# memo
def subsetSumToKMemo(n, k, nums):
    dp = [[-1 for j in range(k+1)] for i in range(n)]
    def solve(i,k):
        if k==0:
            return True
        if i==0:
            return False
        if dp[i][k] != -1:
            return dp[i][k]
        flag1, flag2 =  False, False
        # don't pick 
        flag1 = solve(i-1,k)
        if k-nums[i] >= 0:
            flag2 = solve(i-1,k-nums[i])
        dp[i][k] = flag1 or flag2
        return dp[i][k]

    return solve(n-1,k)

# tabulation
def subsetSumToKTab(n, k, arr):
    dp = [[False for j in range(k+1)] for i in range(n)]
    for i in range(n):
        dp[i][0] = True
    if arr[0] <= k:
        dp[0][arr[0]] = True
    for i in range(1,n):
        for j in range(1,k+1):
            dp[i][j] = dp[i-1][j] 
            if arr[i] <= j:
                dp[i][j] = dp[i][j] or dp[i-1][j-arr[i]]
    return dp[n-1][k]

# space optimization
def subsetSumToK(n, k, arr):
    prev, curr = [False] * (k+1), [False] * (k+1)
    prev[0],curr[0] = True,True
    
    if arr[0] <= k:
        prev[arr[0]] = True
    
    for i in range(1,n):
        for j in range(1,k+1):
            curr[j] = prev[j] 
            if arr[i] <= j:
                curr[j] = curr[j] or prev[j-arr[i]]
        # reuse same arrays
        for j in range(1,k+1):
            prev[j] = curr[j]
            curr[j] = False

    return prev[k]
```

##### 14. Partition Equal Subset Sum 

```python
class Solution:
    def subsetSumToK(self, n, k, arr):
        prev, curr = [False] * (k+1), [False] * (k+1)
        prev[0],curr[0] = True,True
        if arr[0] <= k:
            prev[arr[0]] = True
        for i in range(1,n):
            for j in range(1,k+1):
                curr[j] = prev[j] 
                if arr[i] <= j:
                    curr[j] = curr[j] or prev[j-arr[i]]
            # reuse same arrays
            for j in range(1,k+1):
                prev[j] = curr[j]
                curr[j] = False
        return prev[k]
    def canPartition(self, nums: List[int]) -> bool:
        s = sum(nums)
        if s & 1: 
            return False
        return self.subsetSumToK(len(nums),s//2,nums)
```

##### 15. Partition Set Into 2 Subsets With Min Absolute Sum Diff 

```python
def minSubsetSumDifference(arr: List[str], n: int) -> int:
    n, k = len(arr), sum(arr)
    dp = [[False for j in range(k+1)] for i in range(n)]

    for i in range(n):
        dp[i][0] = True
        
    if arr[0] <= k:
        dp[0][arr[0]] = True

    for i in range(1,n):
        for j in range(1,k+1):
            # don't pick
            dp[i][j] = dp[i-1][j]
            # pick
            if arr[i] <= j:
                dp[i][j] = dp[i][j] or dp[i-1][j-arr[i]]
    ans = 1e9
    for j in range(0,k+1):
        if dp[n-1][j] == True:
            ans = min(ans, abs((k-j)-j))
    return ans
```

##### 16. Count Subsets with Sum K

```python
def findWays(nums: List[int], k: int) -> int:
    n = len(nums)
    # dp[n][k+1]
    dp = [[-1 for j in range(k+1)] for i in range(n)]
    
    def solve(i,k):
        # base cond
        if i==0:
            # [0,1,....]
            if k==0:
                if nums[0] == 0:
                    return 2
                else:
                    return 1
            elif nums[0] == k: 
                return 1
            return 0
            
        if dp[i][k] != -1:
            return dp[i][k]
        # pick 
        x, y = 0,0
        if nums[i] <= k: 
            x = solve(i-1,k-nums[i])
        # don't pick
        y = solve(i-1,k)
        dp[i][k] = x + y
        return x + y

    return solve(n-1,k)
```

```python
def findWaysDP(nums: List[int], k: int) -> int:
    n = len(nums)
    # dp[n][k+1]
    dp = [[0 for j in range(k+1)] for i in range(n)]

    # init
    for i in range(n): # k == 0
        dp[i][0] = 1
    if nums[0] == 0:
        dp[0][0] = 2
    elif nums[0] <= k:
        dp[0][nums[0]] = 1

    # for loop
    for i in range(1,n):
        for j in range(1,k+1):
            x,y = 0,0
            if nums[i] <= j: 
                x = dp[i-1][j-nums[i]]
            y = dp[i-1][j]
            dp[i][j] = x + y
    return dp[n-1][k]
```

##### 18. 0/1 Knapsack

```python
def knapsack(n,w,wi,vi):
    dp = [[-1 for j in range(w+1)] for i in range(n)]
    
    def solveMemo(i,w):
        if i==0:
            if wi[0] <= w: return vi[0]
            return 0

        if dp[i][w] != -1:
            return dp[i][w]

        x,y = 0,0
        # don't pick
        x = solveMemo(i-1,w)
        # pick
        if w - wi[i] >= 0: 
            y = solveMemo(i-1,w-wi[i]) + vi[i]
        dp[i][w] =max(x,max(y,-1))
        return dp[i][w]

    def solveDP(n,w):
        dp2 = [[0 for j in range(w+1)] for i in range(n)]
        for j in range(wi[0],w+1):
            dp2[0][j] = vi[0]
        for i in range(1,n):
            for j in range(w+1):
                x,y = dp2[i-1][j], 0
                if wi[i] <= j:
                    y = dp2[i-1][j-wi[i]] + vi[i]
                dp2[i][j] = max(x,y)
        return dp2[n-1][w]
        
    return solveDP(n,w)
```

##### 19. Minimum Coins

```python
def coinChange(self, coins: List[int], k: int) -> int:
    n = len(coins)
    dp = [[-1 for j in range(k+1)] for i in range(n)]
    def solve(i,k):
        if i==0:
            if k % coins[0] == 0: 
                return k//coins[0]
            else:
                return 1e9
        if dp[i][k] != -1:
            return dp[i][k]
        x,y = 1e9,1e9
        # pick 
        if coins[i] <= k:
            y = solve(i,k-coins[i]) + 1
        x = solve(i-1,k)
        dp[i][k] = min(x,y)
        return dp[i][k]

    ans = solve(n-1,k)
    if ans >= 1e9:
        return -1
    return ans
```

##### 20. Target Sum

```python
class Solution:
    def findTargetSumWays(self, nums: List[int], target: int) -> int:
        # trick ===> {subset1} - {subset2}  == target
        # partition the array with difference = target
        # s1 + s2 = s
        # s1 - s2 = target
        # s2 = (s - target)//2 --> to avoid out of bounds
        n,s = len(nums),sum(nums)
    
        
        def solveDP(n,k):
            dp = [[0 for j in range(k+1)] for i in range(n)]
            if nums[0] == 0: 
                dp[0][0] = 2
            else: 
                dp[0][0] = 1
                
            if nums[0] != 0 and nums[0] <= k:
                dp[0][nums[0]] = 1

            for i in range(1,n):
                for j in range(k+1):
                    x,y = 0,0
                    x = dp[i-1][j]
                    if nums[i] <= j:
                        y = dp[i-1][j-nums[i]]
                    dp[i][j] = x + y
            return dp[n-1][k]

        if (s - target) % 2 or (s - target) < 0 : 
            return 0
        return solveDP(n,(s - target)//2)
```

##### 21. Coin Change 2 

```python
def change(self, k: int, coins: List[int]) -> int:
    n = len(coins)
    dp = [[-1 for j in range(k+1)] for i in range(n)]
    def solve(i,k):
        if i==0:
            if k % coins[0] == 0: 
                # just use one coin since we need to count number of ways
                return 1
            return 0
        if dp[i][k] != -1:
            return dp[i][k]
        x,y = 0,0
        # don't pick
        x = solve(i-1,k)
        # pick
        if coins[i] <= k:
            y = solve(i,k-coins[i])
        dp[i][k] = x + y
        return dp[i][k]

    return solve(n-1,k)
```

##### 22. Unbounded Knapsack 

```python
def unboundedKnapsack(n: int, w: int, vi: List[int], wi: List[int]) -> int:
    dp = [[-1 for j in range(w+1)] for i in range(n)]
    
    def solveMemo(i,w):
        if i==0:
            if wi[0] <= w: 
                # since multiple copies allowed 
                return vi[0] * (w//wi[0])
            return 0

        if dp[i][w] != -1:
            return dp[i][w]

        x,y = 0,0
        # don't pick
        x = solveMemo(i-1,w)
        # pick
        if w - wi[i] >= 0: 
            y = solveMemo(i,w-wi[i]) + vi[i]
        dp[i][w] =max(x,max(y,-1))
        return dp[i][w]
    
    return solveMemo(n-1,w)
```

##### 23. Rod Cutting Problem

```python
def cutRod(vi, n):
    wi = [i for i in range(1,n+1)]
    dp = [[-1 for j in range(n+1)] for i in range(n)]

    def solve(i,n):
        if i==0:
            return n * vi[0]
        if dp[i][n] != -1:
            return dp[i][n]
        x,y = 0,0
        # don't pick
        x = solve(i-1,n)
        # pick
        if wi[i] <= n:
            y = solve(i,n-wi[i]) + vi[i]
        dp[i][n] = max(x,y)
        return dp[i][n]
    
    return solve(n-1,n)
```

```python
def maxProfit(prices) -> int:
    # pick min and calculate the profit
    low_price, max_profit = 1e9,0
    for price in prices:
        low_price = min(low_price, price)
        max_profit = max(max_profit, price - low_price)
    return max_profit
```

##### 357. Count Numbers with Unique Digits

```python
def countNumbersWithUniqueDigits(n: int) -> int:
    if n==0: return 1
    if n==1: return 10
    def solve(n):
        ans,c = 9, 9
        for i in range(1,n):
            ans *= c
            c -= 1
        return ans
    ans = 10
    for j in range(2,n+1):
        ans += solve(j)
    return ans
```
