##### 2. Level Order

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right

def levelOrder(root):
    queue,level_order = [],[]
    if root is None:
        return level_order
    queue.append(root)
    while len(queue) != 0:
        size,temp = len(queue),[]
        for i in range(0,size):
            node = queue[0]
            del queue[0]
            if node:
                temp.append(node.val)
                if node.left is not None:
                    queue.append(node.left)
                if node.right is not None:
                    queue.append(node.right)
        level_order.append(temp)
    return level_order 
```

##### 8. Height of a Binary Tree

```python
import queue
class Solution:
    def maxDepth(self, root: Optional[TreeNode]) -> int:
        # bfs (or) level order
        if root is None:
            return 0
        def levelorder(root):
            q, depth = queue.Queue(), 0
            q.put(root)
            while q.empty() is not True:
                qsize = len(q.queue)
                for i in range(qsize):
                    node = q.get()
                    if node is None:
                        continue
                    if node:
                        if node.left:
                            q.put(node.left)
                        if node.right:
                            q.put(node.right)
                depth += 1
            return depth
        return levelorder(root)
```

##### 10.  Diameter of Binary Tree

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def diameterOfBinaryTree(self, root) -> int:
        ans = 0
        if root is None:
            return 0
        def solve(root):
            if root is None:
                return 0
            if root:
                left_height = solve(root.left)
                right_height = solve(root.right)
                nonlocal ans
                ans = max(ans,left_height + right_height)
                return max(left_height,right_height) + 1
        solve(root)
        return ans
```

##### 11.  Binary Tree Maximum Path Sum

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def maxPathSum(self, root) -> int:
        ans = 0
        def solve(root):
            nonlocal ans
            if root is None:
                return 0
            if root:
                left_value=max(0,solve(root.left))  # don't carry -ve
                right_value=max(0,solve(root.right))  # don't carry -ve
                ans = max(ans,left_value + right_value + root.val)
                return max(left_value,right_value) + root.val
        solve(root)
        return ans
```

##### 13. Binary Tree Zigzag Level Order Traversal

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
import queue
class Solution:
    def zigzagLevelOrder(self, root):
        if root is None:
            return []
        def levelOrder(root):
            queue,level_order = [],[]
            if root is None:
                return level_order
            queue.append(root)
            flag = True
            while len(queue) != 0:
                size,temp = len(queue),[]
                for i in range(0,size):
                    node = queue[0]
                    del queue[0]
                    if node:
                        temp.append(node.val)
                        if node.left is not None:
                            queue.append(node.left)
                        if node.right is not None:
                            queue.append(node.right)
                if flag: 
                    level_order.append(temp)
                else:
                    level_order.append(temp[::-1])
                flag = not flag
            return level_order 
        return levelOrder(root)

                            
```

##### 19. Symmetric Binary Tree

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def isSymmetric(self, root):
        
        def solve(root1,root2):
            if root1 is None and root2 is None:
                return True
            if root1 is None or root2 is None:
                return False
            if root1.val!= root2.val:
                return False
            return solve(root1.left,root2.right) and solve(root1.right,root2.left)
        if root is None:
            return True
        return solve(root.left,root.right)
```

##### 21. Lowest Common Ancestor of a Binary Tree

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        def solve(root,p,q):
            if root is None:
                return None
            if root.val == p.val or root.val == q.val:
                return root
            l = solve(root.left,p,q)
            r = solve(root.right,p,q)
            if l is None and r is None:
                return None
            elif l is None and r is not None:
                return r
            elif r is None and l is not None:
                return l
            else:
                return root
        return solve(root,p,q)
```

##### 22. Maximum width of a Binary Tree

```python
# index the nodes starting from 1


# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
import queue
class Solution:
    def widthOfBinaryTree(root):
        q = queue.Queue()
        q.put((root,1))
        ans = 0
        while len(q.queue) != 0:
            size = len(q.queue)
            for i in range(size):
                node, idx = q.get()
                if node:
                    if i==0:
                        left_most = idx
                    if i==size-1:
                        right_most = idx
                    if node.left:
                        q.put((node.left,idx*2))
                    if node.right:
                        q.put((node.right,idx*2+1))
            ans = max(ans,right_most-left_most+1)
        return ans

            
```

##### 24.  All Nodes Distance K in Binary Tree

```python
# graph bfs idea, but here we need parent pointers

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None
import queue

def distanceK(root, target, k):
    # bread first search in three directions: left, right, parent
    # we need parents 
    parent = {}
    vis,ans = [False] * 501,[]

    def get_parents(root):
        q = queue.Queue()
        q.put(root)
        parent[root] = root
        if root.left:
            q.put(root.left)
            parent[root.left] = root
        if root.right:
            q.put(root.right)
            parent[root.right] = root

        while len(q.queue) != 0:
            node = q.get()
            if node.left:
                q.put(node.left)
                parent[node.left] = node
            if node.right:
                q.put(node.right)
                parent[node.right] = node

    get_parents(root)

    def solve(node,dist):
        if node is None:
            return None
        if dist is None:
            return 
        if vis[node.val]:
            return 
        if dist == 0:
            ans.append(node.val)
        else:
            vis[node.val] = True
            solve(node.left,dist-1)
            solve(node.right,dist-1)
            solve(parent[node],dist-1)
    solve(target,k)
    return ans
```

###### 26. Count Complete Tree Nodes

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right

def countNodes(root) -> int:
    # if left_height == right_height, #nodes = 2<<left_height-1
    def getHeightLeft(root):
        count = 0
        while root.left:
            count += 1
            root = root.left
        return count

    def getHeightRight(root):
        count = 0
        while root.right:
            count += 1
            root = root.right
        return count

    def solve(root):
        if root is None:
            return 0
        left_height = getHeightLeft(root)
        right_height = getHeightRight(root)
        if left_height == right_height:
            return (2 << left_height) - 1
        return 1 + solve(root.left) + solve(root.right)

    return solve(root)
```

##### 28. Construct Binary Tree from Preorder and Inorder Traversal

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right

def buildTree(self, preorder, inorder):
    inorder_mapper = {} # to store the indices
    n = len(inorder)
    for i in range(n):
        inorder_mapper[inorder[i]] = i

    def solve(preorder,p_s,p_e,inorder,i_s,i_e):
        if p_s > p_e or i_s > i_e: 
            return None
        root = TreeNode(preorder[p_s])
        idx = inorder_mapper[root.val]
        left = idx - i_s
        root.left = solve(preorder,p_s+1,p_s+left,inorder,i_s,idx-1)
        root.right = solve(preorder,p_s+left+1,p_e,inorder,idx+1,i_e)
        return root
    return solve(preorder,0,n-1,inorder,0,n-1)
```

##### 29. Construct Binary Tree from Inorder and Postorder Traversal

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right

def buildTree(self, inorder, postorder):
    inorder_mapper = {}
    n = len(inorder)
    for i in range(n):
        inorder_mapper[inorder[i]] = i

    def solve(inorder,i_s,i_e,postorder,p_s,p_e):
        if i_s > i_e or p_s > p_e:
            return None

        root = TreeNode(postorder[p_e])
        idx = inorder_mapper[root.val]
        left = idx - i_s

        root.left = solve(inorder,i_s,idx-1,postorder,p_s,p_s+left-1)
        root.right = solve(inorder,idx+1,i_e,postorder,p_s+left,p_e-1)
        return root
    return solve(inorder,0,n-1,postorder,0,n-1)
```
