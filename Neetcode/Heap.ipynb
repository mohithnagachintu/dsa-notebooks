**How heap parent/child relationships work in an array (vector):**

For an element at index i in the array:

Parent index = (i - 1) / 2 (integer division)

Left child index = 2 * i + 1

Right child index = 2 * i + 2

**What should your comparator return?**

Return true if the first argument has lower priority than the second.

Return false if the first argument has higher or equal priority.

##### 1.Kth Largest Element in a Stream

```python
class KthLargest:

    def __init__(self, k: int, nums: List[int]):
        self.min_heap = nums
        self.k = k
        heapq.heapify(self.min_heap)
        while len(self.min_heap) > k:
            heapq.heappop(self.min_heap)
    
    def add(self, val: int) -> int:
        heapq.heappush(self.min_heap, val)
        if len(self.min_heap) > self.k:
            heapq.heappop(self.min_heap)
        return self.min_heap[0]
```

##### 2.Last Stone Weight

```python

def lastStoneWeight(self, stones: List[int]) -> int:
    if len(stones) == 1:
        return stones[0]

    neg_stones = [-s for s in stones]
    heapq.heapify(neg_stones)

    while len(neg_stones) > 1:
        a = heapq.heappop(neg_stones)
        b = heapq.heappop(neg_stones)
        
        if a == b:
            pass
        else:
            heapq.heappush(neg_stones, a-b)

    return abs(neg_stones[0]) if len(neg_stones) else 0

```

##### 3. pickGifts

```python

def pickGifts(self, pos_gifts: List[int], k: int) -> int:
    gifts = [-g for g in pos_gifts]
    heapq.heapify(gifts)
    for i in range(k):
        num = heapq.heappop(gifts)
        heapq.heappush(gifts, -math.floor(math.sqrt(-num)))

    ans = 0
    while gifts:
        ans += heapq.heappop(gifts)
    return -ans
```

```python
# class Solution {
# public:
#     struct CustomCompare{
#         bool operator()(pair<int,int> &a,pair<int,int> &b){
#             return a.second > b.second;
#         }
#     };
#     int findLeastNumOfUniqueInts(vector<int>& arr, int k) {
#         map<int,int> mp;
#         for (int i=0;i<arr.size();i++)
#             mp[arr[i]] += 1;

#         priority_queue<pair<int,int>, vector<pair<int,int>>, CustomCompare> pq;
#         for (auto it: mp){
#             pq.push({it.first, it.second});
#         }

#         while (k > 0 && !pq.empty()){
#             pair<int,int> top = pq.top();
#             pq.pop();
#             if (top.second <= k){
#                 k -= top.second;
#                 mp[top.first] = 0;
#             } else {
#                 pq.push({top.first, top.second - k});
#                 k = 0;
#             }
#         }
#         return pq.size();
#     }
# };
class Solution:
    def findLeastNumOfUniqueInts(self, arr: List[int], k: int) -> int:
        mapper = defaultdict(int)
        for num in arr:
            mapper[num] += 1
        heap = [(v,k) for k,v in mapper.items()]
        heapq.heapify(heap)

        while k > 0 and len(heap):
            count, num = heapq.heappop(heap)
            if (count <= k):
                k -= count
            else:
                k = 0
                heapq.heappush(heap, (count-k,num))
        return len(heap)
```

##### 6.Maximum Subsequence Score

```python
class Solution:
    def maxScore(self, nums1: List[int], nums2: List[int], k: int) -> int:
        n = len(nums1)
        pairs = []
        for i in range(n):
            pairs.append((nums2[i],nums1[i]))
        pairs.sort(key = lambda x : -x[0])
        
        heap = [x[1] for x in pairs[:k]]
        s = sum(heap)
        heapq.heapify(heap)

        ans = s * pairs[k-1][0]

        for j in range(k,n):
            s -= heapq.heappop(heap)
            s += pairs[j][1]
            heapq.heappush(heap, pairs[j][1])

            ans = max(ans, s * pairs[j][0])

        return ans
```
