##### 1106. Parsing A Boolean Expression

```python
class Solution:
    def evaluate(self, operator, values):
        if operator == '!':
            return 't' if values[0] == 'f' else 'f'
        elif operator == '|':
            temp = False 
            for v in values:
                temp = temp or (True if v == 't' else False)
            return 't' if temp else 'f'
        elif operator == '&':
            temp = True  
            for v in values:
                temp = temp and (True if v == 't' else False)
            return 't' if temp else 'f' 
        else:
            return None 

    def parseBoolExpr(self, expression: str) -> bool:
        stack = []
        for ch in expression:
            if ch == 't' or ch == 'f' or ch == '|' or ch == '&' or ch == '(' or ch == '!':
                stack.append(ch)
            elif ch == ',':
                pass
            elif ch == ')':
                values = []
                while stack[-1] != '(':
                    values.append(stack[-1])
                    stack.pop()
                stack.pop() # pops (
                operator = stack.pop()
                result = self.evaluate(operator, values)
                stack.append(result)
            # print(stack)
        return True if stack[-1] == 't' else False
sol = Solution()
print(sol.parseBoolExpr("&(|(f))"))
print(sol.parseBoolExpr("|(f,f,f,t)"))
print(sol.parseBoolExpr("!(&(f,t))"))
```
