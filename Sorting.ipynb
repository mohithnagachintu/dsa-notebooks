##### 1. Bubble sort

```python
def bubble_sort(arr):
    """
    Repeatedly swap adjacent elements if they are in the wrong order. Largest element "bubbles" to the end after each pass.
    Stable - Yes
    Inplace - Yes
    TC - O(N**2)
    SC - O(1)
    """
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(n-i-1):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = arr[j+1],arr[j]
                print(arr)
                swapped = True
        print('----')
        if not swapped:
            # all sorted - early breaking point
            break
    return arr

    
# nums = [5,2,3,1]
# bubble_sort(nums)
```

##### 2. Selection sort

```python
def selection_sort(arr):
    """
    Find minimum element and place it at the correct position.   
    Stable - No
    Inplace - Yes
    TC - O(N**2) (Always)
    SC - O(1)
    Pros: Minimum no. of swaps
    """
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1,n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i],arr[min_idx] = arr[min_idx], arr[i]
        print(arr)
        print('----')
    return arr

# nums = [5,2,3,1]
# selection_sort(nums)
```

##### 3. Insertion Sort

```python
def insertion_sort(arr):
    """
    Insert each element into its correct position in the already sorted left part.   
    Stable - Yes
    Inplace - Yes
    TC - O(N**2) 
    SC - O(1)
    Pros: Excellent for nearly sorted array
    """

    n = len(arr)
    for i in range(1,n):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j+1] = arr[j]
            j -= 1
            print(arr)
        arr[j+1] = key
        print(arr)
        print('---')
    return arr

# nums = [5,2,3,1]
# insertion_sort(nums)
```

##### 4. Merge Sort

```python
def merge_sort(arr,low,high):
    """
    Divide array into halves, sort both recursively, then merge.
    Stable - Yes
    Inplace - No
    TC - O(N log N) 
    SC - O(N)
    """

    def merge(left, right):
        result = []
        i = j = 0
        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
        while i < len(left):
            result.append(left[i])
            i += 1
        while j < len(right):
            result.append(right[j])
            j += 1
        return result
    
    length = high - low +  1
    if length <= 1:
        return arr[low:high+1]
    mid = (low + high) // 2
    left = merge_sort(arr,low,mid)
    right = merge_sort(arr,mid+1,high)
    return merge(left,right)

nums = [5,2,3,1]
# merge_sort(nums,0,len(nums)-1)
```

##### 5. Quick sort

```python
def quick_sort(arr, low, high):
    """
    Choose pivot and partition array around it. Lomuto with two pointers.
    Stable No
    In-place Yes
    TC - Average: O(N log N)
    TC - Worst: O(N²)
    """
    
    def partition(arr, low, high):
        pivot = arr[high]
        i = low
        j = high - 1
        while True:
            while i <= j and arr[i] <= pivot:
                i += 1
            while i <= j and arr[j] > pivot:
                j -= 1
            if i >= j:
                break
            arr[i], arr[j] = arr[j], arr[i]
        # moving the pivot into its final sorted position
        arr[i], arr[high] = arr[high], arr[i]
        return i

    if low >= high:
        return

    p = partition(arr, low, high)
    quick_sort(arr, low, p - 1)
    quick_sort(arr, p + 1, high)

nums = [5,2,3,1]
quick_sort(nums,0,len(nums)-1)
print(nums)
```
