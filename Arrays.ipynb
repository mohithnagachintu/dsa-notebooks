##### 1. Two Sum

```python
def twoSum(nums, target):
    mp = dict()
    n = len(nums)
    ans = [-1,-1]
    for i in range(n):
        j = mp.get(target-nums[i])
        if j is not None:
            ans[0] = i
            ans[1] = j
        mp[nums[i]] = i
    return ans
```

##### 2. Three Sum

```python
# T(N) = O(NlogN) + O(N^2)
def threeSum(nums):
    n = len(nums)
    nums.sort()
    ans = []
    i = 0
    while i < n:
        target = 0-nums[i]
        # do a binary search
        low,high = i+1,n-1
        while low < high:
            s = nums[low] + nums[high]
            if s == target:
                temp = [nums[i],nums[low],nums[high]]
                ans.append(temp)
                
                # remove duplicates of nums[low]
                while low < high and nums[low] == temp[1]:
                    low += 1
                    
                # remove duplicates of nums[high]
                while low < high and nums[high] == temp[2]:
                    high -= 1
                    
            elif s < target:
                low += 1
            elif s > target:
                high -= 1
                
        # remove duplicates of nums[i]
        while i + 1 < n and nums[i] == nums[i+1]:
            i += 1
        i += 1
    return ans
```
