##### a. (Prerequisite to next one - Cyclic Sort [1...N]

Cyclic Sort is a very useful algorithm when:

The array contains numbers from 1 to N
Each number is supposed to be at its correct index
Often used in problems like:
- Missing Number
- Find All Missing Numbers
- Find Duplicate Number
- Find Corrupt Pair

```python
def cyclic_sort(nums):
    i = 0
    while i < len(nums):
        correct_index = nums[i] - 1

        if nums[i] != nums[correct_index]:
            nums[i], nums[correct_index] = nums[correct_index], nums[i]
        else:
            i += 1
    print(nums)
    return nums

# worst case
nums = [2,3,4,5,1]
print(cyclic_sort(nums))
```

##### 1. First Missing Positive

```python
class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        # Approach 1 - Using visited array
        # TC -> O(N)
        # SC -> O(N)
        # n = len(nums)
        # vis = [False] * (n+1)
        # for i in range(n):
        #     if 0 <= nums[i] <= n:
        #         vis[nums[i]] = True
        # for i in range(1,n+1):
        #     if not vis[i]:
        #         return i
        # # [1,2,3,4,5] -> return 6 (n+1)
        # return n + 1

        # Approach 2 -> Cyclic Sort - Without extra space
        # TC - O(2 * N))
        # SC - O(1)
        n = len(nums)
        i = 0
        while i < n:
            correct_index = nums[i] - 1
            if 0 < nums[i] <= n and nums[i] != nums[correct_index]:
                # swap
                temp = nums[i]
                nums[i] = nums[correct_index]
                nums[correct_index] = temp
            else:
                # i in correct position
                i += 1
            
        for i in range(n):
            if nums[i] != i + 1:
                return i + 1
        return n + 1
```

##### 2. Text Justification

```python
class Solution:
    def fullJustify(self, words: List[str], maxWidth: int) -> List[str]:
        def get_words(i):
            line = []
            length = 0
            while i < len(words) and length + len(words[i]) <= maxWidth:
                line.append(words[i])
                length += len(words[i]) + 1
                i += 1
            return line
        
        def create_line(line,i):
            length = 0
            for word in line:
                length += len(word) + 1 # for mandatory space
            length -= 1  # remove last space

            pending_space = maxWidth - length
            if len(line) == 1 or i == len(words):
                return ' '.join(line) +  " " * pending_space
            
            word_count = len(line) - 1
            spaces_per_word = pending_space // word_count
            left_extra_spaces = pending_space % word_count
            for j in range(left_extra_spaces):
                line[j] += ' '
            for j in range(word_count):
                line[j] += ' ' * spaces_per_word
            return ' '.join(line)
        
        i = 0
        ans = []
        while i < len(words):
            line = get_words(i)
            i += len(line)
            ans.append(create_line(line,i))
        return ans
```
