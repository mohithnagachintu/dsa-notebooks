##### 1. Dynamic Range Sum Queries - CSES

```python
import sys


class FenwickTree:
    def __init__(self, n):
        self.bit = [0] * (n + 1)
        self.n = n

    def update(self, i, delta):
        while i <= self.n:
            self.bit[i] += delta
            i += (i & -i)

    def query(self, i):
        total = 0
        while i > 0:
            total += self.bit[i]
            i -= (i & -i)
        return total

    def range_query(self, l, r):
        return self.query(r) - self.query(l - 1)


# data = list(map(int, sys.stdin.buffer.read().split()))
# n, q = data[0], data[1]
# arr = [0] + data[2 : 2 + n]
# fenwick_tree = FenwickTree(n)

# for i in range(1, n + 1):
#     fenwick_tree.update(i, arr[i])

# ans = []
# index = 2 + n
# for _ in range(q):
#     op, l, r = data[index], data[index + 1], data[index + 2]
#     index += 3

#     if op == 1:
#         fenwick_tree.update(l, r - arr[l])
#         arr[l] = r
#     else:
#         ans.append(str(fenwick_tree.range_query(l, r)))

# print("\n".join(ans))
```

##### 2. Range Sum Query - Mutable (Leetcode - 307)

```python
class FenwickTree:
    def __init__(self, n):
        self.bit = [0] * (n + 1)
        self.n = n

    def update(self, i, delta):
        while i <= self.n:
            self.bit[i] += delta
            i += (i & -i)

    def query(self, i):
        total = 0
        while i > 0:
            total += self.bit[i]
            i -= (i & -i)
        return total

    def range_query(self, l, r):
        return self.query(r) - self.query(l - 1)

class NumArray:

    def __init__(self, nums: list[int]):
        self.nums = nums
        self.fenwick_tree = FenwickTree(len(nums))
        for i in range(len(nums)):
            self.fenwick_tree.update(i+1, nums[i])
        
    def update(self, index: int, val: int) -> None:
        self.fenwick_tree.update(index + 1, val - self.nums[index])
        self.nums[index] = val

    def sumRange(self, left: int, right: int) -> int:
        return self.fenwick_tree.range_query(left + 1, right + 1)
```
