# Fast I/O in C++: `ios::sync_with_stdio(false)` and `cin.tie(nullptr)`

## 1. Why does C++ I/O need optimization?

By default, C++ input/output (`cin`, `cout`) is **slow** because:

1.  C++ streams are **synchronized with C stdio** (`scanf`, `printf`)
2.  `cin` is **tied to `cout`**, causing unnecessary flushing

This safety comes at the cost of performance, which matters in
competitive programming (CSES, Codeforces).

------------------------------------------------------------------------

## 2. `ios::sync_with_stdio(false);`

### What it does

Disables synchronization between: - C++ streams (`cin`, `cout`) - C
stdio (`scanf`, `printf`)

### Effect

-   `cin` / `cout` become much faster
-   Comparable to `scanf` / `printf`

### Important rule

After disabling sync:

❌ **Do NOT mix**

``` cpp
cin / cout
```

with

``` c
scanf / printf
```

Doing so leads to **undefined behavior**.

------------------------------------------------------------------------

## 3. `cin.tie(nullptr);`

### What does "tie" mean?

By default: - `cin` is *tied* to `cout` - Before every `cin`,
`cout.flush()` is called

This ensures prompts appear before input in interactive programs.

### Why it's bad for CP

-   `flush()` is expensive
-   Completely unnecessary for batch input problems

### What this line does

``` cpp
cin.tie(nullptr);
```

-   Unties `cin` from `cout`
-   Prevents automatic flushing
-   Speeds up input

------------------------------------------------------------------------

## 4. Using both together

``` cpp
ios::sync_with_stdio(false);
cin.tie(nullptr);
```

### Result

-   🚀 2--5× faster I/O
-   Standard practice in competitive programming

------------------------------------------------------------------------

## 5. When should you use this?

  Scenario               Use it?
  ---------------------- ---------------
  Small inputs           Optional
  CSES                   ✅ Yes
  Codeforces             ✅ Yes
  Google interview       ⚠️ Optional
  Interactive problems   ❌ Be careful

------------------------------------------------------------------------

## 6. Common mistake ❌

``` cpp
ios::sync_with_stdio(false);
cin.tie(nullptr);

printf("Hello");
cin >> x;
```

This is **undefined behavior**.

Always choose **one I/O system**: - Either `cin / cout` - Or
`scanf / printf`

------------------------------------------------------------------------

## 7. Interview-ready explanation

> "`ios::sync_with_stdio(false)` disables synchronization between C and
> C++ streams to improve performance, and `cin.tie(nullptr)` prevents
> unnecessary flushing of output before input operations."

This is a **strong, concise answer**.

------------------------------------------------------------------------

## 8. Takeaway

-   Not always required, but **good habit**
-   Should be written **by default** in competitive programming
-   Essential for large inputs

------------------------------------------------------------------------

**Recommendation:**\
Always add these two lines at the start of `main()` unless writing an
interactive problem.


# Sweep Line Sweep Line Algorithm

### What it is
- A technique to process **intervals or events on a line (usually time or coordinates)** by converting them into **start/end events**.
- Events are sorted, and a running state (counter, sum, active set) is updated while sweeping from left to right.

### Core idea
- Convert each interval `[l, r]` into:
  - `(l, +1)` → start of activity
  - `(r, -1)` → end of activity
- Sort events by position (and tie-break carefully).
- Maintain a running value (e.g., active intervals).
- Track maximum / minimum / transitions as required.

### Time & Space Complexity
- **Time:** `O(n log n)` due to sorting
- **Space:** `O(n)` for event storage
- Efficient and deterministic (preferred in CP).

---

## When to Use Sweep Line

### Use it when:
- You need to find **maximum overlaps** (e.g., max customers, max meetings).
- You are working with **intervals** and care about how many are active at a time.
- Events can be modeled as **add/remove** actions on a timeline.
- The problem involves **ranges, segments, or time windows**.

### Common CSES / Interview problems
- Maximum number of overlapping intervals
- Minimum number of rooms/resources
- Range add/remove with queries
- Detecting conflicts or peak usage

### Important subtlety
- **Tie-breaking matters**:
  - Decide whether “end” happens before “start” at the same position.
  - This depends on whether intervals are inclusive or exclusive.


# Optimizing `unordered_map` Performance in C++

## 1. `reserve(n)`

### What it does

``` cpp
mp.reserve(n);
```

-   Pre-allocates space for **at least `n` elements**
-   Prevents repeated **rehashing** while inserting elements

### What is rehashing?

-   When the hash table grows, it:
    -   Allocates a larger bucket array
    -   Re-inserts **all existing elements**
-   This operation is expensive (linear in current size)

### Why `reserve` matters

Without `reserve`: - Multiple reallocations - Multiple rehashes - Slower
performance for large inputs

With `reserve(n)`: - Buckets allocated once - Insertions stay amortized
**O(1)** - Faster and more predictable performance

**Rule:**\
If you know the number of elements in advance, always call `reserve()`.

------------------------------------------------------------------------

## 2. `max_load_factor(x)`

### What is load factor?

    load_factor = number_of_elements / number_of_buckets

Example: - 70 elements - 100 buckets\
→ load factor = `0.7`

### Why load factor matters

-   Higher load factor ⇒ more collisions
-   More collisions ⇒ slower lookups
-   Extreme collisions ⇒ near O(n) behavior

### Default behavior

-   Default load factor ≈ `1.0`
-   Allows table to become fairly dense

### What this line does

``` cpp
mp.max_load_factor(0.7);
```

-   Forces rehashing earlier
-   Keeps buckets less crowded
-   Reduces collisions
-   Improves lookup speed

Trade-off: - Slightly higher memory usage - Much better performance

------------------------------------------------------------------------

## 3. Why use `reserve` and `max_load_factor` together?

``` cpp
mp.reserve(n);
mp.max_load_factor(0.7);
```

Together they: - Allocate enough buckets upfront - Maintain low bucket
density - Minimize collisions - Avoid costly rehash storms

This makes `unordered_map`: \> Fast, stable, and suitable for large
inputs

------------------------------------------------------------------------

## 4. When should you use these?

### Use them when:

-   `n ≥ 10^5`
-   Input size is known beforehand
-   Performance is critical
-   Using `unordered_map` in competitive programming

### Not needed when:

-   Input size is small
-   Hash map usage is minimal
-   Performance is not a concern

------------------------------------------------------------------------

## 5. Interview-ready explanation

> "We use `reserve` to avoid rehashing and `max_load_factor` to reduce
> collisions, ensuring stable constant-time performance for hash table
> operations."

------------------------------------------------------------------------

## 6. Important note

-   These optimizations improve **expected** performance
-   They do **not** guarantee worst-case O(1)
-   Worst-case behavior is still O(n), but much harder to hit

------------------------------------------------------------------------

## 7. Key Takeaway

-   `reserve()` → avoids rehashing
-   `max_load_factor()` → reduces collisions
-   Use both for high-performance hashing


# Advanced `unordered_map` Usage Notes (Competitive Programming)

------------------------------------------------------------------------

## 1. Custom Hash using `splitmix64`

### Why a custom hash is needed

-   `unordered_map` can be **hacked** with adversarial inputs
-   Many competitive programming platforms (including CSES) have inputs
    that:
    -   Cause excessive hash collisions
    -   Degrade performance to **O(n²)**

A custom hash makes collisions extremely unlikely.

------------------------------------------------------------------------

### `splitmix64` Hash Function

``` cpp
struct custom_hash {
    static uint64_t splitmix64(uint64_t x) {
        x += 0x9e3779b97f4a7c15;
        x = (x ^ (x >> 30)) * 0xbf58476d1ce4e5b9;
        x = (x ^ (x >> 27)) * 0x94d049bb133111eb;
        return x ^ (x >> 31);
    }

    size_t operator()(uint64_t x) const {
        static const uint64_t FIXED_RANDOM =
            chrono::steady_clock::now().time_since_epoch().count();
        return splitmix64(x + FIXED_RANDOM);
    }
};
```

### How to use it

``` cpp
unordered_map<long long, int, custom_hash> mp;
```

### Benefits

-   Prevents collision-based attacks
-   Makes `unordered_map` safe for large inputs
-   Commonly used in high-level Codeforces / CSES solutions

------------------------------------------------------------------------

## 2. When NOT to Use `unordered_map`

### Avoid `unordered_map` when:

#### ❌ 1. Deterministic performance is required

-   `unordered_map` has **expected O(1)**, not guaranteed
-   Worst case can still be `O(n)`

**Use instead:** - `map` - Sorting + binary search - Arrays / vectors

------------------------------------------------------------------------

#### ❌ 2. You need ordered data

-   `unordered_map` does NOT maintain order

**Use instead:** - `map` - `set` - Sorting-based approaches

------------------------------------------------------------------------

#### ❌ 3. Memory is tight

-   Hash tables use:
    -   Buckets
    -   Pointers
    -   Extra overhead

**Use instead:** - Arrays - Coordinate compression + vector -
Sorting-based solutions

------------------------------------------------------------------------

#### ❌ 4. Small constraints

-   For `n ≤ 1e4`, hash maps are often overkill
-   Sorting or brute-force is simpler and safer

------------------------------------------------------------------------

#### ❌ 5. Problems known to be hash-hostile

-   Some CSES problems are designed to break naive hashing
-   Without a custom hash, `unordered_map` may TLE

**Rule:**\
If using `unordered_map` on large input → use `reserve()` +
`custom_hash`.

------------------------------------------------------------------------

## 3. Summary Table

  Scenario                    Best Choice
  --------------------------- -------------------------------
  Fast average lookup         `unordered_map`
  Deterministic behavior      `map` / sorting
  Interval / range problems   Sweep line / prefix sums
  Large input + hashing       `unordered_map` + custom hash
  Ordered traversal           `map`

------------------------------------------------------------------------

## 4. Key Takeaways

-   Use `unordered_map` for speed, not safety
-   Always optimize it for large inputs
-   Prefer sorting when possible
-   Custom hash = professional CP practice
