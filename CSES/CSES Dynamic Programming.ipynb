# DP Compression Rule (2D → 1D)

## What is DP Compression?
DP compression means reducing the memory usage of a dynamic programming solution
(usually from 2D to 1D) **without changing correctness**.

The key idea is:
> You can overwrite DP states **only if their values are no longer needed**.

---

## The Core Rule (Very Important)

> **DP can be compressed if all dependencies of the current state are already computed
> or safely preserved at the time of overwrite.**

So compression is **not about dimensions**, it is about **dependencies**.

---

## Example: Unbounded Knapsack (Min Coins)

### 2D State Definition
```
dp[i][j] = minimum coins to make sum j using coins i..n-1
```

### Recurrence
```
dp[i][j] = min(
    dp[i+1][j],          // next row, same column
    1 + dp[i][j-a[i]]    // same row, left side
)
```

### Dependency Analysis

| Dependency | Row | Column |
|----------|-----|--------|
| dp[i+1][j] | next row | same |
| dp[i][j-a[i]] | same row | left |

---

## Why Compression Works Here

When we use a single array `dp[j]` and process `j` **left → right**:

- `dp[j]` still represents `dp[i+1][j]` (not overwritten yet)
- `dp[j-a[i]]` already represents `dp[i][j-a[i]]` (updated earlier)

Both dependencies are available at the right time.

✅ Safe to compress.

---

## Correct 1D Loop Order

```
for each coin:
    for j = coin to s:
        dp[j] = min(dp[j], 1 + dp[j - coin])
```

⚠️ Loop order matters.

---

## Contrast: 0/1 Knapsack

### Recurrence
```
dp[i][j] = max(
    dp[i-1][j],
    dp[i-1][j-w[i]] + v[i]
)
```

Both dependencies come from the **previous row**.

### Correct Loop Order
```
for j = s down to w[i]
```

This preserves previous-row values.

---

## Universal Compression Checklist

Before compressing DP, check:

1. Does dp[i][j] depend only on i or i±1?
2. Are same-row dependencies strictly on one side (left or right)?
3. Can loop order guarantee those values are already computed?

If YES to all → compression is safe.

If NO → do NOT compress.

---

## Summary Table

| Problem Type | Dependency | Loop Direction |
|------------|-----------|----------------|
| Unbounded knapsack | same row (left) | left → right |
| Coin change (ways) | same row (left) | left → right |
| 0/1 knapsack | previous row | right → left |
| LIS (O(n²)) | multiple states | not compressible |

---

## One-Line Takeaway

> **DP compression is safe when overwrite order preserves all dependencies.**
