##### 1. Remove Outermost Parentheses

```python
def removeOuterParentheses( s: str) -> str:
    open, st = 0, ""
    for ch in s:
        if ch == '(':
            open += 1
            if open > 1:
                st += "("
        else:
            if open > 1:
                st += ')'
            open -= 1
        # print(ch,open)
    return st
s = "(()())(())"
removeOuterParentheses(s)
```

##### 6. Rotate String

```python
def rotateString(self, s: str, goal: str) -> bool:
    if len(s) != len(goal): 
        return False
    if s in goal + goal:
        return True
    else:
        return False
```

##### 7.  Valid Anagram

```python
def isAnagram(s: str, t: str) -> bool:
    n, m = len(s), len(t)
    if (n!=m):
        return False
    c1, c2 = [0] * 26, [0] * 26
    for i in range(n):
        c1[ord(s[i])-ord('a')] += 1
        c2[ord(t[i])-ord('a')] += 1
    for i in range(26):
        if c1[i] != c2[i]: return False
    return True
```

##### 8. Sort Characters By Frequency

```python
import queue
def frequencySort(s: str) -> str:
    pq = queue.PriorityQueue()
    char_count = dict()
    for ch in s:
        char_count[ch] = char_count.get(ch,0) + 1
    for key in char_count:
        pq.put((char_count[key],key))

    ans = ""
    while pq.empty() == False:
        top = pq.get()
        ans += top[1] * top[0]
    return ans[::-1]
```

##### 9. Maximum Nesting Depth of the Parentheses

```python
def maxDepth(s: str) -> int:
    maxi,open = 0,0
    for ch in s:
        if ch == '(':
            open += 1
        elif ch == ')':
            maxi = max(maxi,open)
            open -= 1
    return maxi
```

##### 10. Roman to Integer

```python
def romanToInt(self, s: str) -> int:
    mp = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
    prev = mp[s[-1]]
    ans = prev
    for i in range(len(s)-2,-1,-1):
        if mp[s[i]] < prev:
            ans -= mp[s[i]]
        else:
            ans += mp[s[i]]
        prev = mp[s[i]]
    return ans
```

##### 15. Minimum Add to Make Parentheses Valid

```python
def minAddToMakeValid(s: str) -> int:
    ans,open = 0,0
    for ch in s:
        if ch == '(':
            open += 1
        else:
            if open == 0:
                ans += 1
            else:
                open -= 1
    return ans + open
```

##### 16. Count and Say (Leetcode - 38)

```python
def countAndSay(self, n: int) -> str:
    def process(st):
        i,n = 0,len(st)
        res = ""
        while i < n:
            ch, c = st[i], 1
            while (i + 1 < n and st[i+1] == ch):
                i+=1
                c+=1
            res += str(c) + ch
            i+=1
        return res
    ans = "1"
    for i in range(1,n):
        ans = process(ans)
    return ans
```
