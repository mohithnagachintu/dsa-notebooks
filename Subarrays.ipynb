##### 1. Subarray sum equals k

```python
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        n = len(nums)
        mapper = dict()
        mapper[0] = 1   # empty prefix subarray - 1

        # direct - [0..j] = prefix_j
        #        - [0..i] = prefix_i
        # if prefix_j -  prefix_i == k ==> [i..j] sum is k
        # indirect way, prefix_j, you check if prefix_i exists using (prefix_j - k)
        
        prefix_sum = 0
        ans = 0
        for i in range(n):
            prefix_sum += nums[i]
            # check 
            if prefix_sum - k in mapper:
                ans += mapper[prefix_sum-k]
            mapper[prefix_sum] = mapper.get(prefix_sum,0) + 1

        return ans
```

##### 2. Continuous Subarray Sum

```python
class Solution:
    def checkSubarraySum(self, nums: List[int], k: int) -> bool:
        # sum[i..j] = k * (Factor) ==> sum[i..j] % k = 0
        # sum[i..j] = (pre[j] - pre[i-1])
        # (pre[j] - pre[i-1])%k = 0
        # pre[j] % k = pre[i-1] % k

        prefix_mapper = dict()
        s = 0
        # to allow a subarray starting from index 0 to have a valid length (e.g., if the sum of the first two elements is $k$, $1 - (-1) = 2$).
        prefix_mapper[0] = -1
        
        for i in range(len(nums)):
            s = (s + nums[i] ) % k
            if s in prefix_mapper:
                idx = prefix_mapper[s]
                if i - idx >= 2:
                    return True
            else:
                prefix_mapper[s] = i
        
        return False
```

##### 4. Longest Subarray with Sum K (GFG) or 325. Maximum Size Subarray Sum Equals k (Leetcode)

```python
class Solution:
    def longestSubarray(self, arr, k):
        n = len(arr)
        mapper = dict()
        
        mapper[0] = -1
        ans = 0
        prefix_sum = 0
        maxi = []
        for i in range(n):
            prefix_sum += arr[i]
            if prefix_sum - k in mapper:
                start_idx = mapper[prefix_sum - k]
                if ans < i - start_idx:
                    ans = i - start_idx
                    maxi = arr[start_idx + 1: i+1]
            if prefix_sum not in mapper:
                mapper[prefix_sum] = i
        # print(maxi)
        return ans
```

##### 5. Subarray Sums Divisible by K


```python
class Solution:
    def subarraysDivByK(self, nums: List[int], k: int) -> int:
                # sum[i..j] % k == 0
        #
        # sum[i..j] = prefix[j] - prefix[i-1]
        #
        # (prefix[j] - prefix[i-1]) % k == 0
        #
        # Therefore:
        # prefix[j] % k == prefix[i-1] % k
        #
        # Let:
        # rem = prefix[j] % k
        #
        # If the same remainder 'rem' has already occurred before at some
        # index p, then:
        #
        # prefix[p] % k == prefix[j] % k
        #
        # => (prefix[j] - prefix[p]) % k == 0
        #
        # => sum[p+1 .. j] % k == 0
        #
        # Therefore, every previous occurrence of the same remainder forms
        # a subarray ending at j whose sum is divisible by k.
        #
        # If rem has occurred 'freq' times before, then we can form
        # 'freq' new valid subarrays ending at j.

        n = len(nums)
        prefix_sum = 0
        mapper = defaultdict(lambda : 0)
        mapper[0] = 1
        ans = 0
        for i in range(n):
            prefix_sum += nums[i]
            target = prefix_sum % k
            if target in mapper:
                ans += mapper[target]
            mapper[target] += 1
        return ans
```
