### 1D DP 

##### 1. Climbing Stars

```java
public int climbStairs(int n) {
    if (n==1) return 1;
    else if (n==2) return 2;
    int dp[] = new int[n+1];
    dp[1] = 1;
    dp[2] = 2;
    for(int i=3;i<=n;i++)
        dp[i] = dp[i-1] + dp[i-2];
    return dp[n];
}
climbStairs(6)
```

##### 2. Frog Jump

```java
public int frogJump(int n, int heights[]) {
    int dp[] = new int[n];
    dp[1] = Math.abs(heights[1]-heights[0]);
    for(int i=2;i<n;i++){
        dp[i] = Math.min(dp[i-1] + Math.abs(heights[i]-heights[i-1]), 
                        dp[i-2] + Math.abs(heights[i]-heights[i-2]));
    }
    return dp[n-1];
}
```

##### 3. Frog Jump with k distances

```java
public int minimizeCost(int n, int k, int []height){
    if (n==1) return 0;

    int dp[] = new int[n];
    dp[1] = Math.abs(height[1]-height[0]);
    for(int i=2;i<n;i++){
        int temp = Integer.MAX_VALUE;
        for (int j=Math.max(0,i-k);j<i;j++){
            temp = Math.min(temp,dp[j] + Math.abs(height[i]-height[j]));
        }
        dp[i] = temp;
    }
    return dp[n-1];
}
int arr[] = {10};
minimizeCost(1,1,arr);
```

##### 4.Maximum sum of non-adjacent elements 

```java
public int rob(int[] nums) {
    // start from 0 or 1
    int n = nums.length;
    if (n==1) return nums[0];
    else if (n==2) return Math.max(nums[0],nums[1]);
    else if (n==3) return Math.max(nums[1],nums[0] + nums[2]);
    else{
        int dp[] = new int[n+1];
        dp[0] = nums[0]; dp[1] = Math.max(nums[0],nums[1]);
        dp[2] = Math.max(nums[1],nums[0] + nums[2]);
        for(int i=3;i<n;i++){
            dp[i] = nums[i] + Math.max(dp[i-2],dp[i-3]);
        }
        dp[n] = Math.max(dp[n-1],dp[n-2]);
        return dp[n];
    }
}
```

##### 5. House Robber

```java
public int rob1(int[] nums,int start,int end) {
    // start from 0 or 1
    int n = end - start + 1;
    if (n==1) 
        return nums[start];
    else if (n==2) 
        return Math.max(nums[start],nums[start+1]);
    else if (n==3) 
        return Math.max(nums[start+1],nums[start] + nums[start + 2]);
    else{
        int dp[] = new int[n+1];
        dp[0] = nums[start]; 
        dp[1] = Math.max(nums[start],nums[start+1]);
        dp[2] = Math.max(nums[start+1],nums[start] + nums[start+2]);

        for (int i=3;i<n;i++){
            dp[i] = nums[start+i] + Math.max(dp[i-2],dp[i-3]);
        }

        dp[n] = Math.max(dp[n-1],dp[n-2]);
        return dp[n];
    }
}
public int rob(int[] nums) {
    int n = nums.length;
    if (n==1) return nums[0];
    return Math.max(rob1(nums,0,n-2),rob1(nums,1,n-1));
}
```
